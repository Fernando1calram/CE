## Ejercicio 2: Búsqueda en Anchura (BFS)

### Contenido teórico trabajado

- **Apartado 1.2:** Formulación del problema (estado inicial, estado objetivo, operadores, restricciones, criterios de evaluación).
- **Apartado 2.1.1:** Búsqueda no informada (ciega) — BFS como algoritmo completo y óptimo en grafos no ponderados.

### Ejemplo resuelto: Encontrar ruta en un grafo

**Problema:** Dado un grafo que representa ciudades conectadas por carreteras, encontrar la ruta más corta (en número de saltos) desde una ciudad origen hasta una ciudad destino usando BFS.

````python
from collections import deque

def bfs(grafo, inicio, objetivo):
    """
    Búsqueda en anchura (BFS) que devuelve la ruta más corta
    desde 'inicio' hasta 'objetivo' en un grafo no ponderado.
    
    Args:
        grafo: dict — grafo de adyacencia {nodo: [vecinos]}.
        inicio: str — nodo de partida.
        objetivo: str — nodo destino.
    
    Returns:
        list — ruta más corta, o None si no existe camino.
    """
    cola = deque([[inicio]])          # Cola de caminos pendientes
    visitados = {inicio}               # Conjunto de nodos ya explorados

    while cola:
        camino = cola.popleft()        # Sacar el camino más antiguo
        nodo = camino[-1]              # Último nodo del camino actual

        if nodo == objetivo:           # ¿Hemos llegado al objetivo?
            return camino

        for vecino in grafo[nodo]:     # Explorar vecinos
            if vecino not in visitados:
                visitados.add(vecino)
                cola.append(camino + [vecino])

    return None                        # No hay camino


# --- Ejemplo de uso ---

grafo_ciudades = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

ruta = bfs(grafo_ciudades, 'A', 'F')
print(f'Ruta: {" -> ".join(ruta)}')
# Salida: Ruta: A -> C -> F
````


**Análisis del ejemplo:**

| Elemento | Equivalente en el código |
|----------|--------------------------|
| **Estado inicial** | `'A'` (nodo de partida) |
| **Estado objetivo** | `'F'` (nodo destino) |
| **Operadores** | Los vecinos de cada nodo (aristas del grafo) |
| **Espacio de estados** | Todos los nodos alcanzables desde `'A'` |
| **Función de coste** | Longitud del camino (número de aristas) |
| **Completo** | Sí — si existe camino, lo encuentra |
| **Óptimo** | Sí — en grafos no ponderados, encuentra el camino más corto |

## Ejercicio 3: Búsqueda en Profundidad (DFS)

### Contenido teórico trabajado

- **Apartado 2.1.1:** DFS como algoritmo de búsqueda en profundidad con backtracking, utilizando una pila LIFO.

### Ejemplo resuelto: Resolver un laberinto simple

**Problema:** Dado un laberinto representado como un grafo, encontrar un camino desde la entrada hasta la salida usando DFS.


````python
def dfs(grafo, inicio, objetivo):
    """
    Búsqueda en profundidad (DFS) iterativa que devuelve un camino
    desde 'inicio' hasta 'objetivo'.
    
    Args:
        grafo: dict — grafo de adyacencia.
        inicio: str — nodo de partida.
        objetivo: str — nodo destino.
    
    Returns:
        list — camino encontrado, o None si no existe.
    """
    pila = [(inicio, [inicio])] # (nodo_actual, camino)
    print(pila[1])
    visitados = {inicio}

    while pila:
        nodo, camino = pila.pop()  # Sacar último añadido (LIFO)

        if nodo == objetivo:
            return camino

        for vecino in grafo[nodo]:
            if vecino not in visitados:
                visitados.add(vecino)
                pila.append((vecino, camino + [vecino]))

    return None


# --- Ejemplo: Laberinto simple ---

laberinto = {
    'ENTRADA': ['A', 'B'],
    'A': ['ENTRADA', 'C'],
    'B': ['ENTRADA', 'D'],
    'C': ['A', 'SALIDA'],
    'D': ['B'],
    'SALIDA': ['C']
}

camino = dfs(laberinto, 'ENTRADA', 'SALIDA')
print(f'Camino: {" -> ".join(camino)}')
# Posible salida: Camino: ENTRADA -> B -> D (no llega a SALIDA)
# O: Camino: ENTRADA -> A -> C -> SALIDA
````



**Análisis del ejemplo:**

| Elemento | Equivalente en el código |
|----------|--------------------------|
| **Estado inicial** | `'ENTRADA'` |
| **Estado objetivo** | `'SALIDA'` |
| **Operadores** | Vecinos de cada nodo |
| **Completo** | No en espacios infinitos (puede perderse en ramas sin salida) |
| **Óptimo** | No — puede devolver un camino largo cuando existe uno más corto |
| **Ventaja** | Usa poca memoria espacial |
