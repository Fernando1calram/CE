## 1. Requisitos Básicos de un Sistema de Resolución de Problemas

En esta unidad vamos a dar un paso fundamental: entender **cómo un sistema informático puede "resolver problemas" de forma autónoma o semiautónoma**. No se trata solo de ejecutar una serie de instrucciones fijas, sino de dotar al sistema de la capacidad de explorar alternativas, evaluar resultados y tomar decisiones inteligentes para alcanzar un objetivo.

Un **sistema de resolución de problemas** es la columna vertebral de muchas aplicaciones de IA que ya utilizamos a diario: desde los navegadores que calculan la ruta más rápida hasta los asistentes virtuales que planifican tareas. Comprender sus requisitos básicos es el primer paso para poder diseñar, clasificar e implementar sistemas inteligentes de forma rigurosa.

### 1.1. Concepto de Sistema de Resolución de Problemas

Un **sistema de resolución de problemas** es un sistema informático diseñado para encontrar soluciones a problemas complejos mediante el uso de algoritmos y modelos de inteligencia artificial.

**Componentes esenciales:**

| Componente | Descripción |
|------------|-------------|
| **Espacio de estados** | Conjunto de todas las situaciones posibles en las que puede encontrarse el sistema. |
| **Estado inicial** | Configuración inicial del problema. |
| **Estado objetivo (meta)** | Configuración que se desea alcanzar. |
| **Operadores** | Acciones que permiten pasar de un estado a otro. |
| **Función de coste** | Métrica para evaluar la calidad de una solución. |
| **Restricciones** | Limitaciones impuestas al sistema (tiempo, recursos, reglas). |

**Ejemplo detallado: Sistema de navegación GPS**

Imagina que usas una aplicación de mapas para ir de tu casa a un destino desconocido. Este es un sistema de resolución de problemas típico. Veamos cómo se mapean los componentes:

| Componente | Aplicación al GPS |
|------------|-------------------|
| **Espacio de estados** | Todos los cruces e intersecciones posibles del mapa. Cada nodo del grafo representa una ubicación concreta. |
| **Estado inicial** | Tu posición actual (coordenadas GPS). |
| **Estado objetivo** | La dirección de destino que has introducido. |
| **Operadores** | Las vías y conexiones disponibles: girar a la izquierda, continuar recto, cambiar de autopista. Cada operador transforma la posición actual en una nueva posición. |
| **Función de coste** | La distancia recorrida o el tiempo estimado de viaje (según la configuración del usuario). El sistema busca minimizar este coste. |
| **Restricciones** | Vías peatonales prohibidas a vehículos, sentido obligatorio en una calle, peajes que deseas evitar, horarios de restricciones de tráfico. |

Este ejemplo muestra que cualquier problema de navegación puede representarse como un **grafo** donde los nodos son ubicaciones y las aristas son las conexiones entre ellas. El sistema de resolución de problemas recorre ese grafo buscando el camino con el menor coste, aplicando los operadores disponibles y respetando las restricciones. De forma análoga, problemas como el **rompecabezas de las torres de Hanoi**, el **juego de 15 piezas** o la **planificación de rutas logísticas** siguen exactamente esta misma estructura conceptual.

> **Nota importante:** No todos los problemas se modelan como búsqueda en un grafo. Algunos requieren razonamiento sobre hechos (sistemas basados en reglas), otros manejan incertidumbre (redes Bayesianas, lógica difusa), y otros aprenden de la experiencia (aprendizaje por refuerzo). La clave es saber identificar qué tipo de modelo es el adecuado.

### 1.2. Formulación del Problema

La **formulación del problema** es la fase más crítica en el diseño de un sistema de resolución de problemas. Un error en esta fase se propaga a lo largo de todo el sistema y puede hacer que la solución encontrada sea incorrecta, ineficiente o directamente inviable. La clave no está en encontrar la solución más rápida, sino en **plantear correctamente el problema** desde el inicio.

El proceso de formulación implica cinco pasos interdependientes que se detallan a continuación:

#### 1. Definir el estado inicial

El estado inicial representa la **situación de partida** del problema. No basta con decir "partimos de aquí": hay que definir con precisión matemática o computacional qué significa ese "aquí". Esto implica:

- **Representación de datos:** ¿Cómo codificamos la situación actual? (vector, grafo, tabla, string...)
- **Validez:** ¿Es este estado factible? (por ejemplo, en un rompecabezas, no puede haber dos piezas en la misma posición).
- **Determinismo:** ¿El estado inicial es único o existe incertidumbre? (si hay incertidumbre, necesitaremos modelos probabilísticos).

> **Pregunta clave:** ¿Existe más de un estado inicial posible? Si es así, ¿debemos considerar todos o podemos partir de un subconjunto?

#### 2. Definir el estado objetivo

El estado objetivo (o **meta**) es la configuración que el sistema debe alcanzar. Aquí conviene distinguir entre:

- **Objetivo exacto:** Una configuración concreta y única (ej.: el tablero del 8-puzzle ordenado).
- **Objetivo parcial (conjunto de metas):** Cualquier estado que cumpla una condición (ej.: "el viajante ha visitado todas las ciudades", sin importar el orden).

> **Pregunta clave:** ¿Es suficiente con alcanzar *cualquier* estado que cumpla la condición, o debemos alcanzar uno específico?

#### 3. Identificar los operadores

Los operadores son las **acciones elementales** que el sistema puede aplicar para transitar de un estado a otro. Su correcta definición es fundamental porque determinan el **espacio de búsqueda** del sistema:

- **Operadores deterministas:** Una acción produce siempre el mismo resultado.
- **Operadores no deterministas:** Una acción puede producir distintos resultados (ej.: lanzar un dado).
- **Operadores condicionales:** Solo pueden aplicarse si se cumple una condición previa (precondición).

> **Pregunta clave:** ¿El conjunto de operadores es finito? ¿Son conmutativos (el orden importa) o no?

#### 4. Definir las restricciones

Las restricciones son las **limitaciones que acotan el espacio de búsqueda**, eliminando soluciones inviables o no deseadas. Pueden ser:

- **Restricciones duras:** No se pueden violar (ej.: un profesor no puede estar en dos aulas simultáneamente).
- **Restricciones blandas:** Se pueden violar penalizando la solución (ej.: preferir que los profesores no den clase los viernes por la tarde).

> **Pregunta clave:** ¿Las restricciones son estáticas (no cambian) o dinámicas (pueden modificarse durante la ejecución)?

#### 5. Establecer criterios de evaluación

No basta con encontrar *una* solución: necesitamos evaluar cuál es **la mejor** entre todas las posibles. Los criterios de evaluación definen la **función de coste** (también llamada función objetivo o *fitness*), que cuantifica la calidad de una solución:

- **Coste de camino:** Suma de costes de las transiciones entre estados.
- **Coste de solución:** Calidad del estado final alcanzado.
- **Múltiples objetivos:** A veces es necesario optimizar varios criterios simultáneamente (ej.: minimizar tiempo y coste).

> **Pregunta clave:** ¿Buscamos la solución óptima (la mejor posible) o una solución suficientemente buena (satisfactoria)?


```mermaid
flowchart LR
    A["1. Estado Inicial"] --> B["2. Operadores"]
    B --> C["3. Estado Objetivo"]
    D["4. Restricciones"] -.-> B
    D -.-> C
    D --> E["5. Criterios de Evaluación"]
```


> **Consejo práctico:** Al formular un problema, es habitual empezar por los extremos (estado inicial y estado objetivo) y luego refinar progresivamente los operadores, restricciones y criterios. Esta aproximación iterativa evita perder tiempo definiendo componentes que luego habrá que modificar.

### 1.3. Ejemplo: Problema de planificación de horarios

**Contexto:** Un centro educativo necesita generar un horario de clases para un trimestre.

| Elemento | Definición |
|----------|------------|
| **Estado inicial** | Lista de profesores, aulas, grupos de alumnos, asignaturas. |
| **Estado objetivo** | Horario completo sin conflictos (aulas, profesores, grupos). |
| **Operadores** | Asignar una asignatura a un profesor en un aula y horario. |
| **Restricciones** | Un profesor no puede estar en dos sitios a la vez; una aula no puede tener dos grupos; horarios laborables. |
| **Criterio de evaluación** | Minimizar huecos, equilibrar carga de trabajo, maximizar ocupación. |

### 1.4. Ejercicio práctico: Formulación de un problema

> **Ejercicio:** Formula un problema de resolución utilizando la estructura anterior:
>
> 1. Elige un problema real (ej.: enrutamiento de vehículos, diagnóstico médico, planificación de producción).
> 2. Define el estado inicial, el estado objetivo, los operadores, las restricciones y los criterios de evaluación.

## 2. Clasificación de Modelos de Inteligencia Artificial

Una vez que sabemos qué es un sistema de resolución de problemas y cómo formularlo, surge una pregunta natural: **¿qué herramientas tenemos para resolverlo?** La respuesta es que existe un amplio abanico de modelos de IA, cada uno diseñado para abordar tipos específicos de problemas. No existe un modelo universalmente mejor: la clave está en **saber elegir el adecuado** para cada contexto.

En este apartado vamos a explorar las principales familias de modelos de IA utilizados en sistemas de resolución de problemas, comprendiendo sus fortalezas, limitaciones y casos de uso. Aprenderemos a:

 - Identificar las principales categorías de modelos de IA y sus características.
 - Distinguir entre modelos de búsqueda, optimización, aprendizaje automático, probabilísticos y otros.
 - Seleccionar el modelo más adecuado para un problema concreto.

A continuación se presenta una tabla de clasificación general de los principales modelos utilizados en sistemas de resolución de problemas:

| Modelo | Enfoque | Tipo de aprendizaje | Aplicación típica |
|--------|---------|---------------------|-------------------|
| **Búsqueda** | Exploración del espacio de estados. | Ninguno (preprogramado) | Juegos, planificación de rutas. |
| **Optimización** | Búsqueda de la mejor solución posible. | Ninguno (preprogramado) | Logística, scheduling. |
| **Aprendizaje Automático (ML)** | Aprendizaje a partir de datos. | Supervisado/No supervisado/Refuerzo | Clasificación, regresión, predicción. |
| **Redes Bayesianas** | Modelado probabilístico de variables. | Supervisado/No supervisado | Diagnóstico, inferencia. |
| **Sistemas Difusos** | Razonamiento con incertidumbre. | Ninguno (preprogramado) | Control de procesos. |
| **Algoritmos Genéticos** | Evolución artificial. | No supervisado | Diseño, optimización. |
| **Redes Neuronales** | Modelado inspirado en el cerebro. | Supervisado/No supervisado/Refuerzo | Reconocimiento de patrones. |

> **Nota importante:** Esta tabla es un **mapa de clasificación**, no una comparativa de rendimiento. Su propósito es dar una visión global de las familias de modelos disponibles. Para comparar modelos de forma ponderada según los requisitos de tu problema, consulta la **Matriz de evaluación ponderada (Apartado 6.2)**, donde se explica cómo asignar pesos a cada criterio.

### 2.1. Modelos de Búsqueda

Los **modelos de búsqueda** constituyen una de las familias más antiguas y fundamentales de la inteligencia artificial. Su idea central es sencilla pero poderosa: **un problema se resuelve explorando sistemáticamente un espacio de estados posibles hasta encontrar una secuencia de acciones que nos lleve desde la situación actual hasta el objetivo deseado**.

Imagina que estás en una ciudad que no conoces y quieres llegar a un museo. No sabes el camino, pero tienes un mapa. Cada intersección es un *estado*, cada calle es un *operador* que te lleva de un estado a otro, y el museo es tu *estado objetivo*. Tu tarea es encontrar la mejor ruta. Eso es exactamente lo que hacen los modelos de búsqueda: explorar un espacio de posibilidades de forma organizada para encontrar la solución.

Estos modelos se dividen en dos grandes categorías según la información que disponen para guiar la búsqueda:

- **Búsqueda no informada (ciega):** No dispone de información adicional sobre qué zonas del espacio de estados son más prometedoras. Explora de forma cíclica y sistemática. Es simple y garantiza encontrar la solución si existe, pero puede ser muy ineficiente en espacios grandes.
- **Búsqueda informada (heurística):** Utiliza información adicional (una *heurística*) para priorizar las ramas más prometedoras del espacio de búsqueda. Es más eficiente, pero no siempre garantiza encontrar la solución óptima.

#### 2.1.1. Búsqueda no informada (ciega)

La **búsqueda no informada** (también llamada *ciega* o *sin información*) explora el espacio de estados sin ninguna pista sobre cuál es la dirección correcta hacia la solución. Esto significa que el algoritmo no sabe si una rama del árbol de búsqueda es más prometedora que otra: avanza de forma sistemática, como un explorador que revisa todas las calles de una ciudad una por una.

A pesar de su aparente simplicidad, estos algoritmos son fundamentales porque **garantizan la completitud** (si existe una solución, la encontrarán) y, en algunos casos, la optimización (la solución encontrada es la mejor posible). Son útiles cuando el espacio de estados es pequeño o cuando no se dispone de información heurística.

| Algoritmo | Descripción | Complejidad espacial | Complejidad temporal |
|-----------|-------------|---------------------|---------------------|
| **BFS (Búsqueda en anchura)** | Explora todos los nodos al mismo nivel antes de pasar al siguiente. Utiliza una cola FIFO. Completo y óptimo en grafos no ponderados. | Alta $O(b^d)$ | Alta $O(b^d)$ |
| **DFS (Búsqueda en profundidad)** | Explora una rama hasta el final antes de retroceder (backtracking). Utiliza una pila LIFO. No es completo en espacios infinitos. | Baja $O(b^m)$ | Alta $O(b^m)$ |
| **Búsqueda por coste uniforme** | Explora siempre el nodo con menor coste acumulado $g(n)$. Completo y óptimo en grafos con costes no negativos. | Alta $O(b^{1+C^*/\epsilon})$ | Alta $O(b^{1+C^*/\epsilon})$ |

> **Notas sobre complejidad:**
> - $b$: factor de ramificación (número medio de hijos por nodo).
> - $d$: profundidad de la solución más superficial.
> - $m$: profundidad máxima del espacio de estados.
> - $C^*$: coste de la solución óptima.
> - $\epsilon$: coste mínimo de una transición.

#### 2.1.2. Búsqueda informada (heurística)

La **búsqueda informada** (o *heurística*) utiliza información adicional para guiar la exploración hacia las zonas más prometedoras del espacio de estados. La clave es la **función heurística** $h(n)$, que estima el coste estimado desde el nodo actual $n$ hasta el estado objetivo. Esta información permite al algoritmo "priorizar" los nodos que parecen más cercanos al objetivo, reduciendo drásticamente el número de nodos explorados.

El concepto de heurística es fundamental: no es una fórmula exacta, sino una **estimación inteligente** que nos dice "probablemente por aquí hay camino". Una heurística bien diseñada puede hacer la diferencia entre resolver un problema en segundos o en años. Sin embargo, si la heurística es mala (o inexistente), la búsqueda informada se degrada hasta comportarse como una búsqueda ciega.

| Algoritmo | Descripción | Heurística | Complejidad espacial | Complejidad temporal |
|-----------|-------------|------------|---------------------|---------------------|
| **A\*** | Combina el coste real acumulado $g(n)$ con una estimación heurística $h(n)$ para priorizar nodos. Es completo y óptimo si la heurística es admisible ($h(n) \leq h^*(n)$). | $f(n) = g(n) + h(n)$ | Alta | Alta |
| **Búsqueda voraz (Greedy)** | Prioriza exclusivamente los nodos que parecen más cercanos al objetivo según la heurística. No considera el coste ya recorrido. Rápido pero no óptimo ni completo. | $f(n) = h(n)$ | Media | Media |
| **Búsqueda local** | Explora vecinos del estado actual y se mueve al que mejor evalúe. Puede quedar atrapado en óptimos locales. No es completo. | $f(n) = h(n)$ (o función de coste) | Baja | Baja |

**Aplicaciones:**
- **Búsqueda en anchura (BFS):** Encontrar la ruta más corta en un grafo no ponderado.
- **Búsqueda en profundidad (DFS):** Resolver laberintos, análisis de archivos.
- **Algoritmo A\*:** Planificación de rutas (GPS, videojuegos).
- **Búsqueda voraz (Greedy):** Problemas donde se necesita una solución rápida pero no óptima.

### 2.2. Modelos de Optimización

Los **modelos de optimización** responden a una pregunta muy práctica: de todas las soluciones posibles a un problema, **¿cuál es la mejor?** Mientras que la búsqueda se centra en encontrar *cualquier* camino válido hasta el objetivo, la optimización va un paso más allá y busca *el camino óptimo* (o una solución lo más cercana posible a la óptima).

Piensa en la logística de una empresa de reparto: hay miles de formas de distribuir los paquetes entre los camiones. Todas son *viables*, pero algunas son mucho más baratas y eficientes que otras. Los modelos de optimización nos ayudan a encontrar esa solución «perfecta» (o al menos, la mejor posible dentro de las restricciones). Se utilizan en ámbitos tan diversos como la planificación de producción, la logística de distribución, el diseño de redes de telecomunicaciones o la asignación de recursos en centros de datos.

#### 2.2.1. Programación lineal

La **programación lineal (PL)** es una de las técnicas de optimización más clásicas y potentes. Su idea es sencilla pero elegante: expresar el problema como una **función objetivo lineal** (por ejemplo, el beneficio total o el coste total) que queremos maximizar o minimizar, sujeta a un conjunto de **restricciones también lineales** (disponibilidad de recursos, capacidades, demandas, etc.).

El método más utilizado para resolver problemas de programación lineal es el **método simplex**, desarrollado por George Dantzig en 1947, que recorre de forma inteligente los vértices del poliedro de soluciones factibles hasta encontrar el óptimo. Además de la PL, existen variantes como la programación entera (cuando las variables deben ser números enteros) o la programación no lineal (cuando la función objetivo o las restricciones no son lineales).

**Objetivo:** Maximizar o minimizar una función lineal sujeta a restricciones lineales.

**Ejemplo:** Maximizar beneficios de producción sujeto a limitaciones de recursos.

#### 2.2.2. Algoritmos genéticos

Los **algoritmos genéticos** son una técnica de optimización inspirada en los principios de la evolución biológica. En lugar de explorar una sola solución a la vez, trabajan con una **población de soluciones candidatas** que evolucionan generación tras generación. La idea central es que las mejores soluciones tienen más probabilidades de "reproducirse" y dar lugar a soluciones aún mejores, mientras que las soluciones deficientes van desapareciendo.

El proceso se inspira directamente en la selección natural: las soluciones se "reproducen" mediante **cruce** (combinar partes de dos soluciones padres para generar hijos) y **mutación** (pequeñas alteraciones aleatorias que introducen diversidad genética). A lo largo de las generaciones, la población tiende a converger hacia soluciones cada vez mejores.

```mermaid
flowchart LR
    A["1. Población<br/>inicial"] --> B["2. Evaluación<br/>fitness"]
    B --> C["3. Selección <br/>de los mejores <br/>individuos"]
    C --> D["4. Cruce"]
    D --> E["5. Mutación"]
    E --> F["6. Nueva<br/>generación"]
    F --> G{"7. ¿Convergencia?"}
    G -- "No" --> B
    G -- "Sí" --> H["Solución"]
```

**Aplicaciones:** Diseño de circuitos, scheduling, optimización de rutas.

#### 2.2.3. Recocido simulado (Simulated Annealing)

El **recocido simulado** es una técnica de optimización inspirada en un proceso de la metalurgia: el recocido. En metalurgia, cuando se calienta un metal y se deja enfriar lentamente, los átomos se reorganizan en una estructura de mínima energía (más estable). El recocido simulado traslada esta idea a la optimización: permite que el algoritmo acepte **soluciones peores temporalmente** con cierta probabilidad, lo que le permite escapar de los óptimos locales y explorar zonas más prometedoras del espacio de búsqueda.

La clave está en la **temperatura**: al inicio, la temperatura es alta y el algoritmo acepta fácilmente soluciones peores (exploración amplia); a medida que la temperatura desciende (enfriamiento), se vuelve más selectivo y se concentra en mejorar la mejor solución encontrada (explotación). La probabilidad de aceptar una solución peor sigue la distribución de Boltzmann: $P(\text{aceptar}) = e^{-\Delta E / T}$, donde $\Delta E$ es el empeoramiento y $T$ es la temperatura actual.

**Principio:** Permite soluciones peores temporalmente para escapar de óptimos locales.

**Aplicaciones:** Problema del viajante (TSP), diseño de circuitos.

### 2.3. Modelos de Aprendizaje Automático (ML)

El **aprendizaje automático** (machine learning) representa un cambio de paradigma respecto a los modelos de búsqueda y optimización: en lugar de que un programador escriba las reglas para resolver un problema, **el sistema aprende las reglas a partir de datos de ejemplo**. Es decir, en lugar de decirle al sistema *cómo* resolver un problema, le mostramos *muchos ejemplos de soluciones* y le pedimos que aprenda a generalizar.

Imagina que quieres construir un sistema que detecte spam en correos electrónicos. Con un modelo de búsqueda, tendrías que escribir manualmente todas las reglas posibles ("si contiene 'Ganador de la lotería', marcar como spam"; "si el remitente es desconocido y contiene un enlace sospechoso", etc.). Con aprendizaje automático, simplemente le proporcionas miles de correos etiquetados como "spam" o "no spam", y el algoritmo aprende patrones que un programador humano probablemente no habría identificado manualmente: combinaciones de palabras, estructuras de frase, patrones de remitente, etc.

Aunque ya hemos visto estos modelos en la Unidad 1, aquí los enfocamos desde la perspectiva de resolución de problemas, destacando cómo cada tipo de aprendizaje se aplica a tareas concretas de automatización y toma de decisiones:

- **Clasificación:** Asignar una categoría a una entrada (ej.: spam/no spam, fraude/legítimo). El modelo aprende a partir de datos etiquetados y generaliza a nuevos casos.
- **Regresión:** Predecir un valor continuo (ej.: precio de una vivienda, demanda de un producto). El modelo aprende la relación entre variables de entrada y una variable numérica de salida.
- **Clustering:** Agrupar datos similares sin etiquetas previas (ej.: segmentación de clientes, detección de anomalías). El algoritmo descubre estructuras ocultas en los datos.
- **Aprendizaje por refuerzo:** Aprender mediante interacción con el entorno (ej.: juego, robótica, recomendación). Un agente aprende a tomar decisiones secuenciales maximizando una recompensa acumulada.

### 2.4. Modelos probabilísticos: Redes Bayesianas

Las **redes Bayesianas** (o redes de creencia) son modelos que permiten razonar bajo **incertidumbre**. A diferencia de los modelos de búsqueda (que asumen un mundo determinista y conocido) o los de aprendizaje automático (que buscan patrones en datos), las redes Bayesianas incorporan explícitamente la **probabilidad** como herramienta para tomar decisiones cuando no disponemos de información completa o cuando los datos son ruidosos.

Imagina un sistema de diagnóstico médico: un paciente presenta fiebre y tos. ¿Tiene gripe? ¿Es coronavirus? ¿Es una alergia? No podemos saberlo con certeza, pero sí podemos calcular **probabilidades** basadas en los síntomas observados. Las redes Bayesianas nos permiten expresar estas relaciones de forma rigurosa: modelar cómo los síntomas dependen de las enfermedades, y cómo las enfermedades dependen entre sí (por ejemplo, si tienes coronavirus, es más probable que tengas fiebre).

Estas redes se construyen mediante un **grafo acíclico dirigido** donde los nodos representan variables aleatorias (síntomas, enfermedades, resultados de pruebas) y las aristas representan dependencias probabilísticas entre ellas. Cada nodo incluye una **tabla de probabilidad condicional (CPT)** que cuantifica cómo se distribuyen sus valores dado el estado de sus nodos padres.

**Definición:** Modelo gráfico probabilístico que representa variables aleatorias y sus dependencias condicionales mediante un grafo acíclico dirigido.

**Componentes:**
- **Nodos:** Variables aleatorias.
- **Arcos:** Dependencias probabilísticas entre variables.
- **Tablas de probabilidad condicional (CPT):** Probabilidades de cada nodo dados sus padres.

**Aplicaciones:**
- Diagnóstico médico: Síntomas → Enfermedad.
- Predicción de fallos: Sensores → Estado del equipo.
- Análisis de riesgo: Factores → Probabilidad de evento.

## 3. Modelos de Automatización de Tareas

Hasta ahora hemos visto cómo los sistemas de IA pueden *resolver problemas* buscando soluciones o *clasificar modelos* para elegir la herramienta adecuada. Pero existe una categoría de aplicaciones de IA que está transformando la forma en que trabajamos a diario: la **automatización de tareas**.

No se trata solo de encontrar la mejor solución a un problema, sino de **hacer que el problema se resuelva solo**, sin intervención humana constante. Imagina un sistema que lee facturas, extrae los datos relevantes, los verifica y los registra en la contabilidad, todo él solo. O un sistema que detecta si una transacción bancaria es fraudulenta y la bloquea antes de que se ejecute. Eso es automatización con IA.

### 3.1. Concepto de automatización de tareas

La **automatización de tareas** mediante IA consiste en utilizar sistemas inteligentes para realizar actividades de forma automática, reduciendo la intervención humana y mejorando la eficiencia. La clave que diferencia esta automatización de la tradicional es la **capacidad de adaptación**: un sistema de automatización con IA puede modificar su comportamiento en función de los datos y el contexto, mientras que un sistema tradicional sigue reglas fijas programadas de antemano.

Para entender bien en qué consiste, pensemos en dos enfoques diferentes de automatizar una tarea:

- **Automatización tradicional (RPA):** Sigue reglas predefinidas y estáticas. Si la entrada se desvía de lo esperado, el sistema falla. Es como un robot de fábrica que solo puede soldar en un punto exacto: si la pieza se mueve medio centímetro, el robot no puede adaptarse.
- **Automatización con IA:** Adapta su comportamiento según los datos y el contexto. Si la entrada varía, el sistema puede ajustarse y seguir funcionando. Es como un operario humano que, aunque las piezas se muevan un poco, puede ajustar su posición para seguir soldando correctamente.

**Diferencias clave:**

| Concepto | Descripción |
|----------|-------------|
| **Automatización tradicional (RPA)** | Sigue reglas predefinidas y estáticas. |
| **Automatización con IA** | Adapta su comportamiento según los datos y el contexto. |

### 3.2. Tipos de automatización con IA

Los sistemas de automatización con IA se pueden clasificar en **tres tipos principales**, según la naturaleza de la tarea que realizan. No son excluyentes: un sistema real puede combinar varios tipos. Por ejemplo, un sistema de gestión de reclamaciones puede *clasificar* la solicitud, *procesar* los documentos necesarios y *tomar una decisión* sobre si aprobarla o rechazarla.

A continuación se presentan los tres tipos, desde el más simple (clasificación) hasta el más complejo (toma de decisiones):

#### 3.2.1. Automatización de clasificación

La **automatización de clasificación** es el tipo más básico y directo de automatización con IA. Su objetivo es **asignar automáticamente una categoría predefinida a un elemento de entrada** (un documento, una imagen, un texto, un sonido). Es como tener un archivista automático que, al recibir un documento, lo lee y lo coloca en el cajón correcto.

Este tipo de automatización se basa fundamentalmente en técnicas de **aprendizaje automático supervisado** (cuando tenemos datos etiquetados) o **no supervisado** (clustering, cuando no tenemos etiquetas previas). Los modelos más utilizados son árboles de decisión, máquinas de soporte vectorial (SVM), redes neuronales y algoritmos de agrupamiento como K-means.

El flujo es sencillo: el sistema recibe un elemento de entrada, lo procesa con un modelo de clasificación y asigna una categoría. La clave está en que el modelo debe ser capaz de generalizar: si ha aprendido a clasificar facturas de una empresa, debe ser capaz de clasificar facturas de otra empresa con un formato ligeramente diferente.

```mermaid
flowchart LR
    A["Documento<br/>de entrada"] --> B["Sistema de<br/>Clasificación"]
    B --> C["Categoría<br/>asignada"]
```

>**Ejemplo: Clasificación automática de documentos en un despacho de abogados**

>Un despacho de abogados recibe cientos de documentos al día: contratos, demandas, escritos, contratos de arrendamiento, poderes notariales, etc. Manualmente, un asistente debe leer cada documento y asignarlo a una carpeta correspondiente. Esto consume tiempo y es propenso a errores.

>Un sistema de automatización de clasificación soluciona esto del siguiente modo:

>1. **Entrada:** El documento se escanea o se recibe en formato digital (PDF, imagen, DOCX).
>2. **Preprocesamiento:** Se aplica OCR (reconocimiento óptico de caracteres) si es una imagen, y se limpia el texto.
>3. **Extracción de características:** Se extraen palabras clave, frases típicas, estructura del documento.
>4. **Clasificación:** Un modelo de aprendizaje supervisado (por ejemplo, una red neuronal o un SVM) asigna una categoría: "Contrato", "Demanda", "Escrito", "Contrato de arrendamiento", etc.
>5. **Archivo automático:** El sistema mueve automáticamente el documento a la carpeta correspondiente en el servidor.

> **Resultado:** Lo que antes requería 30 minutos por documento, ahora se resuelve en segundos, con una precisión del 95% o superior.

#### 3.2.2. Automatización de procesos (RPA + IA)

La **automatización de procesos** va un paso más allá de la clasificación: no se limita a asignar una categoría, sino que **automatiza flujos de trabajo completos** que involucran múltiples pasos, decisiones y sistemas. Es como si, además de clasificar el documento, el sistema lo leyera, extrajera los datos relevantes, los verificara, los registrara en el sistema contable y avisara a las personas implicadas.

Para lograr esto, se combina **RPA (Robotic Process Automation)** — robots de software que simulan acciones humanas en interfaces digitales — con **modelos de inteligencia artificial** que añaden capacidad de comprensión y toma de decisiones. A diferencia de un RPA tradicional (que solo sigue reglas fijas), un sistema de automatización de procesos con IA puede manejar variaciones en las entradas y adaptarse a situaciones inesperadas.

Los componentes principales son:
- **Robots de software:** Simulan acciones humanas (hacer clic, escribir, copiar, pegar).
- **Inteligencia artificial:** Añade comprensión del lenguaje natural, reconocimiento de imágenes, toma de decisiones.
- **Orquestación:** Coordina y gestiona los robots y los flujos de trabajo.

**Aplicaciones:**
- Procesamiento de facturas (OCR + NLP).
- Gestión de reclamaciones.
- Onboarding de nuevos empleados.
- Procesamiento de solicitudes de préstamo.

>**Ejemplo: Procesamiento automatizado de facturas en una empresa de suministros**

>Una empresa de suministros recibe más de 500 facturas al mes de distintos proveedores. Cada factura tiene un formato diferente, con distintos campos, idiomas y calidades de escaneo. Manualmente, un empleado debe abrir cada factura, leer los datos (importe, fecha, proveedor, concepto), introducirlos en el sistema contable y verificar que coinciden con la orden de compra.

>Un sistema de automatización de procesos soluciona esto del siguiente modo:

>1. **Recepción:** Las facturas llegan por correo electrónico o se escanean.
>2. **Extracción de datos (OCR + NLP):** Un sistema de visión por ordenador extrae el texto de la imagen. Un modelo de procesamiento de lenguaje natural (NLP) identifica los campos clave: proveedor, importe, fecha, IVA, concepto.
>3. **Verificación:** El sistema compara los datos extraídos con la orden de compra correspondiente. Si hay una discrepancia (por ejemplo, el importe supera un umbral), se marca para revisión humana.
>4. **Registro contable:** Si todo coincide, el robot de software abre el sistema contable, introduce los datos y guarda el documento como referencia.
>5. **Notificación:** Se envía un correo al responsable indicando que la factura ha sido procesada.

> **Resultado:** Se reduce el tiempo de procesamiento de 15 minutos por factura a menos de 1 minuto, con una reducción del 90% en errores humanos.

#### 3.2.3. Automatización de decisiones

La **automatización de decisiones** es el tipo más avanzado de automatización con IA. Su objetivo no es solo clasificar o procesar, sino **tomar decisiones automáticamente** basadas en datos y modelos predictivos. Es decir, el sistema no solo ejecuta una tarea, sino que decide *qué* hacer.

Imagina un sistema bancario que recibe una transacción de 5.000 € de una cuenta que habitualmente solo mueve 50 €. Un sistema de clasificación simplemente etiquetaría la transacción como "grande". Un sistema de automatización de decisiones va más allá: analiza el historial del cliente, la hora, la ubicación, el patrón habitual de gastos y decide si la transacción es legítima o sospechosa de fraude. Si es sospechosa, puede bloquearla automáticamente y solicitar confirmación al cliente.

Los principales ejemplos incluyen:
- **Scoring crediticio:** Decisión automática de aprobación de préstamos.
- **Detección de fraude:** Bloqueo automático de transacciones sospechosas.
- **Mantenimiento predictivo:** Programación automática de mantenimiento.

>**Ejemplo: Detección de fraude en transacciones bancarias**

>Un banco recibe miles de transacciones diarias. Algunas son legítimas, otras son fraudulentas (robos de identidad, phishing, tarjetas clonadas). Detectar el fraude manualmente sería imposible dada la cantidad de transacciones. Un sistema de automatización de decisiones soluciona esto del siguiente modo:

>1. **Recepción de la transacción:** Se recibe una nueva operación (por ejemplo, una compra de 800 € con una tarjeta de crédito).
>2. **Análisis de contexto:** El sistema analiza múltiples variables: historial del titular (monto habitual, frecuencia, tipo de compras), ubicación geográfica (¿coincide con movimientos anteriores del cliente?), hora del día, dispositivo utilizado.
>3. **Modelo predictivo:** Un modelo de aprendizaje automático (por ejemplo, una red neuronal o un bosque aleatorio) calcula una **probabilidad de fraude** para esa transacción.
>4. **Decisión automática:**
   - Si la probabilidad es baja (< 5%): se aprueba automáticamente.
   - Si la probabilidad es media (5%-30%): se envía una notificación al cliente para confirmación (SMS, app).
   - Si la probabilidad es alta (> 30%): se bloquea la transacción y se alerta al departamento de fraude.
>5. **Retroalimentación:** Cuando el cliente confirma que una transacción bloqueada era legítima, esa información se utiliza para reentrenar el modelo y mejorar su precisión.

> **Resultado:** Se reducen las pérdidas por fraude en un 70-80%, se minimizan las falsas alarmas (que generan fricción con el cliente) y se libera al personal del banco de tareas de revisión manual.

### 3.3. Implementación de un sistema de automatización

La **implementación de un sistema de automatización** con IA no es simplemente «instalar un software»: es un proceso que requiere planificación, diseño y una iteración continua. Un error en cualquiera de las fases puede hacer que el sistema no funcione como se espera o que genere resultados incorrectos.

El proceso se puede dividir en **siete pasos**, que se detallan a continuación. Cada paso tiene sus propios desafíos y requiere la colaboración de distintos perfiles: analistas de negocio, ingenieros de datos, ingenieros de IA y desarrolladores de software.

**Pasos para implementar un sistema de automatización:**

1. **Identificar la tarea:** ¿Qué proceso se va a automatizar? No todas las tareas son candidatas a automatización. Se debe priorizar aquellas que sean repetitivas, voluminosas y propensas a errores humanos.
2. **Analizar el proceso:** ¿Cuáles son los pasos, decisiones y puntos de fallo? Se debe mapear el flujo actual paso a paso, identificando cuellos de botella, errores frecuentes y puntos donde la intervención humana es realmente necesaria.
3. **Seleccionar la técnica:** ¿Qué modelo de IA es adecuado? ¿Se necesita clasificación, optimización, procesamiento de lenguaje natural, visión por ordenador? Esta decisión condicionará todo el diseño posterior.
4. **Recopilar datos:** ¿Qué datos se necesitan para entrenar el modelo? Los datos son el combustible de cualquier sistema de IA. Sin datos suficientes y de calidad, el modelo no puede aprender.
5. **Entrenar el modelo:** ¿Cómo se entrena y valida? Se divide el conjunto de datos en conjuntos de entrenamiento y prueba, se entrena el modelo y se evalúa su rendimiento con métricas adecuadas.
6. **Implementar:** ¿Cómo se integra en el sistema existente? El modelo debe integrarse con los sistemas existentes (bases de datos, APIs, interfaces de usuario) y funcionar en tiempo real o casi real.
7. **Monitorear y mejorar:** ¿Cómo se mide el rendimiento y se actualiza? Un sistema de IA no es «instalar y olvidar». Los datos evolucionan, los patrones cambian y el modelo debe ser reentrenado periódicamente.

> **Consejo práctico:** Es habitual que los primeros intentos de automatización no funcionen a la primera. El proceso es iterativo: se implementa una versión mínima, se mide su rendimiento, se identifican los fallos y se mejora. Esta aproximación ágil reduce el riesgo y permite aprender rápidamente.

## 4. Modelos de Razonamiento Impreciso

Los sistemas de IA que hemos visto hasta ahora — búsqueda, optimización, aprendizaje automático, redes Bayesianas — operan en un mundo de certezas: o algo es verdadero o es falso, o un dato pertenece a una clase o no pertenece. Pero, ¿alguna vez has tenido que tomar una decisión con información incompleta o ambigua? ¿Alguna vez has descrito algo como "un poco frío" o "bastante caliente"? Los humanos no pensamos en verdades absolutas de 0 o 1: pensamos en matices, en grados, en conceptos que no tienen bordes definidos.

### 4.1. Concepto de Razonamiento Impreciso


Los **modelos de razonamiento impreciso** (también conocidos como **lógica difusa** o *fuzzy logic*) fueron propuestos por Lotfi Zadeh en 1965 y responden precisamente a esta necesidad: permitir que un sistema de IA razonara de forma similar a un humano, manejando conceptos vagos e imprecisos. Mientras que la lógica clásica (booleana) opera con valores absolutos — verdadero o falso, 1 o 0 —, la lógica difusa opera con **grados de verdad**, es decir, con valores intermedios entre 0 y 1.

Piensa en cómo describirías la temperatura de un día de primavera: no dirías simplemente "hace calor" o "hace frío". Dirías "hace un poco fresco", "está templado", "hace calorcito". Estos son **conceptos difusos**: no tienen un límite preciso. ¿A partir de cuántos grados un día se convierte en "caliente"? ¿25 °C? ¿28 °C? ¿30 °C? Distintas personas podrían dar respuestas distintas, y todas serían válidas. La lógica difusa captura esta subjetividad de forma matemática.

**Diferencia con la lógica booleana:**

| Lógica Booleana | Lógica Difusa |
|-----------------|----------------|
| Valores: 0 (falso) o 1 (verdadero) | Valores: cualquier número entre 0 y 1 |
| Conceptos: "caliente" o "no caliente" | Conceptos: "un poco caliente", "muy caliente" |
| Sin incertidumbre | Maneja la incertidumbre |
| Binario | Gradual |

### 4.2. Conjuntos Difusos (Fuzzy Sets)

Para que un sistema de IA pueda trabajar con conceptos difusos como "caliente", "frío" o "alto", necesitamos una forma matemática de representarlos. Aquí es donde entran los **conjuntos difusos** (fuzzy sets).

Un **conjunto difuso** es una generalización de un conjunto clásico (el tipo de conjuntos que aprendemos en matemáticas). En un conjunto clásico, un elemento o pertenece o no pertenece: por ejemplo, el número 30 pertenece al conjunto de "números mayores que 25", pero el 20 no. En un conjunto difuso, un elemento puede pertenecer **parcialmente** a un conjunto, con un **grado de pertenencia** que va de 0 (no pertenece en absoluto) a 1 (pertenece completamente), pasando por cualquier valor intermedio.

Volviendo al ejemplo de la temperatura: a 20 °C, ¿cuánto pertenece este valor al conjunto difuso "frío"? Podría ser 0.7 (bastante frío). ¿Y al conjunto "templado"? Podría ser 0.3 (un poco templado). Fíjate en que 20 °C puede pertencer a varios conjuntos difusos a la vez, con distintos grados de pertenencia. Esto es lo que permite a un sistema de lógica difusa manejar la ambigüedad de forma natural.

Cada conjunto difuso se define mediante una **función de pertenencia** $\mu(x)$, que asigna a cada valor $x$ del dominio un grado de pertenencia entre 0 y 1. Las funciones de pertenencia más comunes son:

| Función | Forma | Uso típico |
|---------|-------|------------|
| **Triangular** | Triángulo | Simple, rápido. Ideal cuando no hay mucha incertidumbre. |
| **Trapezoidal** | Trapecio | Rango con zona plana. Útil cuando un rango de valores es igualmente válido. |
| **Gaussiana** | Campana | Suavidad. Ideal cuando los valores centrales son mucho más probables que los extremos. |
| **Sigmoidal** | S | Transición suave entre 0 y 1. Ideal para conceptos unidireccionales ("alto", "bajo"). |

> **Nota importante:** La elección de las funciones de pertenencia es un aspecto clave del diseño de un sistema difuso. No existe una respuesta "correcta" única: depende del dominio y de la experiencia del diseñador. Es habitual empezar con funciones triangulares o trapezoidales por su simplicidad y ajustarlas según los resultados.

### 4.3. Sistema de Inferencia Difusa

Un **sistema de inferencia difusa** es el corazón de cualquier aplicación de lógica difusa. Su función es tomar **entradas numéricas precisas** (valores *crisp*, como 23.5 °C), procesarlas a través de reglas difusas similares al razonamiento humano, y producir una **salida numérica precisa** (como un nivel de potencia del 67%).

El sistema se compone de cuatro etapas secuenciales que funcionan como un flujo de procesamiento. Cada etapa tiene un papel específico y, juntas, permiten que un sistema de IA "razone" de forma similar a un ser humano: con matices, con incertidumbre y con conocimiento experto codificado en reglas IF-THEN.

**Componentes:**

| Componente | Descripción |
|------------|-------------|
| **Fuzzificación** | Conversión de valores crisp (numéricos) a valores difusos. Se aplica la función de pertenencia de cada conjunto difuso a cada valor de entrada. |
| **Base de reglas** | Conjunto de reglas IF-THEN difusas que codifican el conocimiento experto del dominio. |
| **Motor de inferencia** | Aplica las reglas a las entradas difusas, combinando los grados de pertenencia para determinar la fuerza de activación de cada regla. |
| **Defuzzificación** | Conversión de valores difusos (resultados combinados) a un valor crisp (numérico) que se puede usar para tomar una acción concreta. |

```mermaid
flowchart LR
    A["Valores Crisp<br/>(entrada)"] --> B["Fuzzificación<br/>(conjuntos difusos)"]
    B --> C["Motor de Inferencia<br/>(reglas difusas)"]
    C --> D["Defuzzificación<br/>(salida)<br/>(conjuntos difusos)"]
    D --> E["Valores Crisp<br/>(acción)"]
```

**Las cuatro etapas explicadas en detalle:**

1. **Fuzzificación:** Se toman los valores numéricos de entrada y se convierten en grados de pertenencia a cada conjunto difuso. Por ejemplo, si la temperatura es 23 °C, se calcula a qué grado pertenece a los conjuntos "frío", "templado", "caliente", etc. Esto se hace aplicando las funciones de pertenencia correspondientes.

2. **Base de reglas:** Se dispone de un conjunto de reglas IF-THEN que codifican el conocimiento experto del dominio. Por ejemplo: "SI la temperatura es alta Y la humedad es alta ENTONCES la potencia del aire acondicionado es máxima". Estas reglas son las que el experto humano proporciona al diseñar el sistema.

3. **Motor de inferencia:** Se aplican las reglas a las entradas difusas. Para cada regla, se determina su **grado de activación** (qué tan válida es esa regla dada la entrada actual). Las reglas se combinan (mediante operadores difusos como MIN para el AND y MAX para el OR) para producir una salida difusa agregada.

4. **Defuzzificación:** Se convierte la salida difusa agregada de vuelta a un valor numérico preciso (crisp) que se puede usar para tomar una acción concreta. Los métodos más comunes son el **centroide** (calcula el centro de gravedad del área bajo la curva de salida), el **método del máximo** (toma el valor con mayor grado de pertenencia) y el **método del centro de máximos** (promedio de los valores con mayor grado de pertenencia).

## 5. Modelos de Sistemas Basados en Reglas

### 5.1. Concepto de Sistemas Basados en Reglas

Un **sistema basado en reglas** es un modelo de IA que utiliza un conjunto de reglas IF-THEN (producción) para representar el conocimiento de un dominio y tomar decisiones.

**Componentes:**

| Componente | Descripción |
|------------|-------------|
| **Base de conocimientos** | Conjunto de reglas IF-THEN que codifican el conocimiento del dominio. |
| **Base de hechos (memoria de trabajo)** | Conjunto de datos actuales del problema. |
| **Motor de inferencia** | Componente que aplica las reglas a los hechos para derivar nuevas conclusiones. |
| **Mecanismo de resolución de conflictos** | Estrategia para seleccionar qué regla aplicar cuando varias son compatibles. |

```mermaid
flowchart TD
    subgraph M["MOTOR DE INFERENCIA"]
        direction LR
        A["Base de hechos<br/>(memoria de trabajo)"]
        B["Base de conocimientos<br/>(reglas IF-THEN)"]
    end

    A --> C["Resolución de<br/>conflictos"]
    B --> C
    C --> D["Nuevos hechos<br/>/ conclusiones"]
```

### 5.2. Estructura de una regla de producción

```
IF <condición> THEN <acción/conclusión> [PRECISIÓN <valor>]
```

**Ejemplo:**

```
IF temperatura > 30 AND humedad > 80
THEN riesgo_calor_alto = SI
PRECISIÓN 0.9
```

### 5.3. Tipos de motores de inferencia

#### 5.3.1. Cadena hacia adelante (Forward Chaining)

**Principio:** Parte de los hechos disponibles y aplica reglas para derivar nuevas conclusiones hasta alcanzar el objetivo.

**Flujo:**
```
Hechos iniciales → Aplicar reglas → Nuevos hechos → Repetir → Objetivo alcanzado
```

**Características:**
- **Dirigido por los datos:** Los datos impulsan el proceso.
- **Adecuado para:** Monitorización, diagnóstico, planificación.
- **Ventaja:** Encuentra todas las conclusiones posibles.
- **Desventaja:** Puede generar muchas conclusiones innecesarias.

**Ejemplo:** Sistema de diagnóstico médico.
```
Hechos: paciente tiene fiebre + dolor de cabeza
Regla 1: IF fiebre AND dolor_cabeza THEN posible_gripe
Regla 2: IF posible_gripe THEN recomendar_descanso
Conclusión: recomendar_descanso
```

#### 5.3.2. Cadena hacia atrás (Backward Chaining)

**Principio:** Parte del objetivo (hipótesis) y verifica si los hechos lo apoyan.

**Flujo:**
```
Objetivo → Buscar reglas que conduzcan al objetivo → Verificar condiciones → Repetir → Objetivo verificado o no
```

**Características:**
- **Dirigido por el objetivo:** El objetivo impulsa el proceso.
- **Adecuado para:** Diagnóstico, planificación, verificación.
- **Ventaja:** Eficiente cuando se busca una conclusión específica.
- **Desventaja:** Puede no encontrar todas las conclusiones posibles.

**Ejemplo:** Sistema de diagnóstico médico.
```
Objetivo: ¿Tiene gripe el paciente?
Regla: IF fiebre AND dolor_cabeza THEN gripe
Verificar: ¿Tiene fiebre? → Sí
Verificar: ¿Tiene dolor de cabeza? → Sí
Conclusión: Sí, tiene gripe
```

### 5.4. Estrategias de resolución de conflictos

Cuando varias reglas son aplicables, se necesita una estrategia para elegir cuál aplicar:

| Estrategia | Descripción |
|------------|-------------|
| **Especificidad** | Aplicar la regla más específica (con más condiciones). |
| **Recencia** | Aplicar la regla más recientemente añadida. |
| **Prioridad** | Asignar prioridad a cada regla. |
| **Orden de aparición** | Aplicar la primera regla compatible en la base de conocimientos. |

### 5.5. Implementación de un sistema basado en reglas

**Pasos para implementar un sistema basado en reglas:**

1. **Identificar el dominio:** ¿En qué área se va a aplicar?
2. **Recopilar conocimiento:** Obtener reglas de expertos del dominio.
3. **Formalizar las reglas:** Convertir reglas naturales a formato IF-THEN.
4. **Construir la base de conocimientos:** Almacenar las reglas.
5. **Construir la base de hechos:** Definir los datos iniciales.
6. **Implementar el motor de inferencia:** Forward o backward chaining.
7. **Implementar la resolución de conflictos:** Elegir estrategia.
8. **Probar y validar:** Verificar con casos de prueba.

### 5.6. Ejemplo: Sistema de detección de fallos en una red

**Dominio:** Detección de fallos en una red de computadoras.

**Base de conocimientos (reglas):**

| ID | Regla |
|----|-------|
| R1 | IF ping_falla AND ip_accesible THEN posible_problema_dns |
| R2 | IF ping_falla AND ip_no_accesible THEN posible_cable_dañado |
| R3 | IF ip_accesible AND puerto_80_cerrado THEN posible_servidor_caído |
| R4 | IF ip_accesible AND puerto_443_cerrado THEN posible_certificado_vencido |
| R5 | IF posible_cable_dañado THEN recomendar_revisar_cableado |
| R6 | IF posible_servidor_caído THEN recomendar_reiniciar_servidor |

**Base de hechos:**
- ping_falla = Verdadero
- ip_accesible = Falso
- puerto_80_cerrado = Desconocido
- puerto_443_cerrado = Desconocido

**Ejecución (Forward Chaining):**
1. Regla R2 se cumple: `posible_cable_dañado = Verdadero`
2. Regla R5 se cumple: `recomendar_revisar_cableado = Verdadero`
3. No hay más reglas aplicables.
4. **Conclusión:** Revisar cableado.

## 6. Valoración de la Adecuación de los Modelos

### 6.1. Criterios de evaluación de modelos

Al implementar un sistema de resolución de problemas, es necesario evaluar la adecuación del modelo seleccionado. Los criterios principales son:

| Criterio | Descripción |
|----------|-------------|
| **Precisión** | ¿Qué tan准确的 son las soluciones? |
| **Robustez** | ¿Cómo se comporta ante datos incompletos o ruidosos? |
| **Escalabilidad** | ¿Puede manejar grandes volúmenes de datos o problemas complejos? |
| **Tiempo de respuesta** | ¿Cuánto tarda en producir una solución? |
| **Coste computacional** | ¿Cuántos recursos (CPU, memoria) necesita? |
| **Mantenibilidad** | ¿Qué facilidad tiene para ser actualizado o modificado? |
| **Explicabilidad** | ¿Se puede explicar por qué el modelo tomó una decisión? |
| **Facilidad de implementación** | ¿Qué dificultad tiene para implementar? |

### 6.2. Matriz de evaluación ponderada

La siguiente tabla muestra una **evaluación cualitativa** de los modelos según criterios genéricos. Sin embargo, en la práctica, la elección del modelo adecuado depende de los requisitos específicos de cada problema. Por eso, esta tabla se utiliza como **herramienta de selección** junto con un sistema de ponderación: asignas pesos a cada criterio según la importancia para tu caso concreto y calculas una puntuación final.

| Modelo | Precisión | Robustez | Escalabilidad | Tiempo | Coste | Explicabilidad | Implementación |
|--------|:---------:|:--------:|:-------------:|:------:|:-----:|:--------------:|:--------------:|
| **Búsqueda** | Media-Alta | Baja | Media | Media | Baja | Alta | Media |
| **Optimización** | Alta | Media | Alta | Alta | Alta | Media | Alta |
| **ML (Supervisado)** | Alta | Media | Alta | Media | Media | Baja | Media |
| **ML (No supervisado)** | Media | Alta | Alta | Media | Media | Baja | Media |
| **Redes Bayesianas** | Media-Alta | Alta | Media | Media | Media | Alta | Alta |
| **Sistemas Difusos** | Media | Alta | Media | Baja | Baja | Alta | Media |
| **Sistemas Basados en Reglas** | Media | Media | Baja | Baja | Baja | Alta | Baja |
| **Algoritmos Genéticos** | Media-Alta | Alta | Alta | Alta | Alta | Baja | Alta |

> **Cómo usar esta matriz:**
> 1. Asigna un peso (0-1) a cada criterio según su importancia para tu problema (ej.: precisión = 0.3, explicabilidad = 0.4, robustez = 0.3).
> 2. Convierte las calificaciones cualitativas en valores numéricos: Baja=3, Media=5, Media-Alta=7, Alta=9.
> 3. Multiplica cada valor por su peso correspondiente.
> 4. Suma los resultados para obtener la puntuación total de cada modelo.
> 5. Selecciona el modelo con mayor puntuación.
>
> Un ejemplo completo se muestra en el **Apartado 6.4**.

### 6.3. Proceso de valoración de adecuación

**Pasos para valorar la adecuación de un modelo:**

1. **Definir los requisitos del problema:** ¿Qué se necesita resolver?
2. **Identificar los criterios de evaluación:** ¿Qué es más importante? (precisión, velocidad, coste, etc.)
3. **Evaluar cada modelo candidato:** Aplicar los criterios a cada modelo.
4. **Ponderar los criterios:** Asignar importancia relativa a cada criterio.
5. **Calcular la puntuación total:** Sumar ponderada de cada criterio.
6. **Seleccionar el modelo con mayor puntuación.**
7. **Validar con casos de prueba.**

### 6.4. Ejemplo: Valoración para un sistema de diagnóstico médico

**Requisitos:**
- Alta precisión (vida o muerte).
- Explicabilidad (médico debe entender la decisión).
- Funcionar con datos incompletos.

| Modelo | Precisión (30%) | Explicabilidad (40%) | Robustez (30%) | **Puntuación** |
|--------|:---------------:|:--------------------:|:--------------:|:--------------:|
| **Redes Bayesianas** | 8 × 0.3 = 2.4 | 9 × 0.4 = 3.6 | 8 × 0.3 = 2.4 | **8.4** |
| **Sistemas Basados en Reglas** | 7 × 0.3 = 2.1 | 10 × 0.4 = 4.0 | 6 × 0.3 = 1.8 | **7.9** |
| **ML (Redes Neuronales)** | 9 × 0.3 = 2.7 | 3 × 0.4 = 1.2 | 7 × 0.3 = 2.1 | **6.0** |
| **Sistemas Difusos** | 6 × 0.3 = 1.8 | 8 × 0.4 = 3.2 | 9 × 0.3 = 2.7 | **7.7** |

**Selección:** Redes Bayesianas (mayor puntuación).

### 6.5. Sistemas híbridos

En muchos casos, ningún modelo individual es suficiente. La solución es combinar varios modelos en un **sistema híbrido**.

**Ejemplos de sistemas híbridos:**

| Sistema Híbrido | Modelos combinados | Beneficio |
|-----------------|-------------------|-----------|
| **Reglas + ML** | Motor de reglas + Clasificador ML | Explicabilidad + Precisión |
| **Búsqueda + ML** | Algoritmo de búsqueda + Modelo predictivo | Eficiencia + Calidad de solución |
| **Difuso + Reglas** | Lógica difusa + Motor de reglas | Manejo de incertidumbre + Precisión |
| **Bayesiano + Redes Neuronales** | Red Bayesiana + Deep Learning | Probabilidad + Reconocimiento de patrones |

## 7. Glosario

| Término | Definición |
|---------|------------|
| **Espacio de estados** | Conjunto de todas las situaciones posibles en las que puede encontrarse un sistema. |
| **Estado objetivo** | Configuración que se desea alcanzar en un problema. |
| **Operador** | Acción que permite pasar de un estado a otro. |
| **Heurística** | Regla práctica que ayuda a encontrar soluciones de forma más eficiente. |
| **Búsqueda en anchura (BFS)** | Algoritmo que explora todos los nodos al mismo nivel antes de pasar al siguiente. |
| **Búsqueda en profundidad (DFS)** | Algoritmo que explora una rama hasta el final antes de retroceder. |
| **Algoritmo A*** | Algoritmo de búsqueda que utiliza una función de coste real + estimación. |
| **Algoritmo genético** | Algoritmo de optimización inspirado en la evolución natural. |
| **Red Bayesiana** | Modelo probabilístico que representa variables y sus dependencias. |
| **Lógica difusa** | Lógica que maneja valores continuos entre 0 y 1. |
| **Conjunto difuso** | Conjunto donde la pertenencia tiene un grado entre 0 y 1. |
| **Fuzzificación** | Proceso de convertir valores numéricos en valores difusos. |
| **Defuzzificación** | Proceso de convertir valores difusos en valores numéricos. |
| **Motor de inferencia** | Componente que aplica reglas a hechos para derivar conclusiones. |
| **Forward chaining** | Cadena hacia adelante: parte de los hechos y deriva conclusiones. |
| **Backward chaining** | Cadena hacia atrás: parte del objetivo y verifica condiciones. |
| **Sistema híbrido** | Sistema que combina múltiples modelos de IA. |
