## 1. Fundamentos de los Sistemas Inteligentes

### 1.1. Introducción

> Imagina que puedes preguntar a tu ordenador: *"¿Cuáles son los productos más vendidos este trimestre y qué recomendaciones me das para mejorar las ventas?"* y que, en lugar de recibir una tabla estática, obtengas una respuesta contextualizada, con gráficos, explicaciones y acciones concretas. Eso es un sistema de inteligencia artificial aplicado a un negocio. La IA no es ciencia ficción; ya está transformando hospitales, bancos, fábricas, escuelas y comercios. En esta unidad aprenderás a reconocer qué es un sistema inteligente, cómo se construye y por qué es una herramienta clave para mejorar la eficiencia operativa de cualquier organización.

#### 1.1.1 Breve historia de la IA

La historia de la Inteligencia Artificial es un recorrido fascinante por intentos, avances, decepciones y renacimientos que nos han traído hasta el punto actual. Comprender este recorrido no es solo un ejercicio histórico: nos ayuda a entender por qué la IA funciona como funciona hoy, qué lecciones hemos aprendido de los fracasos del pasado y por qué la revolución actual es diferente a las anteriores.

Todo comenzó con una pregunta fundamental: **¿Pueden las máquinas pensar?** En 1950, el matemático británico Alan Turing propuso el famoso "Test de Turing" como forma de responder a esta pregunta de manera operativa: si un evaluador humano no puede distinguir si está interactuando con una máquina o con un ser humano, la máquina puede considerarse inteligente. Esta idea sentó las bases de lo que hoy llamamos IA.

En 1956, en la histórica Conferencia de Dartmouth, John McCarthy acuñó por primera vez el término "Inteligencia Artificial". Los primeros años fueron de entusiasmo: se construyeron los primeros sistemas expertos como DENDAL (para razonamiento químico) y MYCIN (para diagnóstico médico). Sin embargo, pronto se toparon con limitaciones computacionales y la falta de datos suficientes, lo que provocó el primer "Invierno de la IA" a finales de los 80.

El verdadero punto de inflexión llegó en 2012, cuando el modelo AlexNet ganó el concurso ImageNet utilizando redes neuronales profundas y GPUs, demostrando que el *Deep Learning* podía superar ampliamente a los métodos clásicos. Desde entonces, la IA ha experimentado un crecimiento exponencial, culminando en la era actual de la IA generativa (2022+), con modelos como GPT-4, Stable Diffusion y Midjourney.

| Periodo | Hito | Descripción |
|---------|------|-------------|
| **1950** | Test de Turing | Alan Turing propone si una máquina puede demostrar inteligencia indistinguible de un humano. |
| **1956** | Conferencia de Dartmouth | Se acuña el término "Inteligencia Artificial" (John McCarthy). |
| **1960-70** | Primeros sistemas expertos | DENDAL (química), MYCIN (medicina). |
| **1980-85** | Boom de las máquinas inteligentes | Hardware especializado para IA. |
| **1987-93** | Primer "Invierno de la IA" | Falta de resultados prácticos y sobrecostes. |
| **1990-2010** | IA estadística y ML | Dominio de enfoques basados en datos sobre los simbólicos. |
| **2012** | Revolución del Deep Learning | AlexNet gana ImageNet usando GPUs y redes profundas. |
| **2017** | Arquitectura Transformer | Google publica "Attention Is All You Need" (base de los LLM). |
| **2022+** | IA Generativa | Explosión de modelos como GPT-4, Stable Diffusion, Midjourney. |

#### 1.1.2. Inteligencia fuerte vs. débil

Uno de los conceptos que generan más confusión en el ámbito de la IA es la distinción entre "inteligencia fuerte" y "inteligencia débil". Esta clasificación, propuesta por el filósofo John Searle en 1980, es fundamental para entender qué podemos esperar realmente de los sistemas actuales y cuáles son las aspiraciones a largo plazo.

**Inteligencia débil (Narrow AI o IA estrecha):** Es un sistema diseñado para realizar una tarea específica con un alto nivel de desempeño. No "entiende" el mundo a nivel general; simplemente es muy bueno en lo que hace. Piensa en un sistema que puede diagnosticar cáncer en radiografías con una precisión superior a la de un radiólogo humano: es extraordinario en esa tarea concreta, pero no puede conducir un coche, escribir un poema ni mantener una conversación. La IA que usamos a diario —desde los filtros de Instagram hasta los algoritmos de recomendación de Spotify— es toda IA débil.

**Inteligencia fuerte (AGI — Artificial General Intelligence):** Sería un sistema con capacidad cognitiva comparable a la de un ser humano en cualquier ámbito: razonamiento abstracto, aprendizaje rápido, transferencia de conocimiento entre dominios, comprensión del lenguaje natural, empatía, creatividad, etc. Un sistema de AGI no se limitaría a una tarea; podría aprender a hacer cualquier cosa que un humano pueda aprender.

| Concepto | Definición | Estado actual |
|----------|------------|---------------|
| **IA Débil (Narrow AI)** | Sistemas diseñados para una tarea específica (ej.: reconocimiento facial, traducción). | **Existente y ampliamente utilizada.** |
| **IA Fuerte (AGI)** | Sistema con capacidad cognitiva igual o superior a la humana en cualquier tarea. | **Teórica. No existe actualmente.** |

> **Nota:** Toda la IA actual es "débil" o "estrecha" (Narrow AI). La AGI sigue siendo un objetivo a largo plazo y un tema de debate científico y filosófico.

### 1.2. Definición de Sistema Inteligente

Un **sistema inteligente** es un sistema capaz de percibir su entorno, procesar la información recibida, tomar decisiones y actuar sobre el entorno para alcanzar unos objetivos predefinidos, todo ello con un grado de autonomía y adaptabilidad.

Para entenderlo mejor, pensemos en un sistema inteligente como un **ciclo continuo de interacción con el mundo**:  

```mermaid
flowchart LR
    A[PERCIBE] --> B[PROCESA]
    B --> C[DECIDE]
    C --> D[ACTÚA]
    D --> E[APRENDE]
    E -.-> B
```

Este ciclo se repite una y otra vez, y cada iteración puede mejorar el resultado de la siguiente.

**Características fundamentales:**

- **Percepción:** Es el "sentido" del sistema. Un sistema inteligente debe ser capaz de captar información del entorno. Esto puede hacerse mediante sensores físicos (cámaras, micrófonos, sensores de temperatura), datos digitales (bases de datos, APIs, feeds en tiempo real) o cualquier otra fuente de información. *Ejemplo:* un sistema de diagnóstico médico "percibe" imágenes de radiografías, análisis de sangre y el historial clínico del paciente.

- **Procesamiento:** Es el "cerebro" del sistema. La información percibida se analiza, interpreta y transforma mediante algoritmos, modelos matemáticos y bases de conocimiento. *Ejemplo:* un sistema de recomendación de Netflix procesa tus historiales de visualización, valoraciones y patrones de comportamiento para entender tus preferencias.

- **Toma de decisiones:** Es el momento en que el sistema elige una acción concreta entre varias posibilidades, basándose en los objetivos que se le han asignado. *Ejemplo:* un sistema de trading algorítmico decide comprar o vender acciones en función de patrones de mercado que ha identificado.

- **Acción:** Es la ejecución de la decisión tomada. El sistema actúa sobre el entorno para producir un cambio. *Ejemplo:* un robot en una fábrica de coches suelda las piezas de la carrocería en la posición exacta que el sistema ha calculado.

- **Adaptabilidad:** Un sistema verdaderamente inteligente no se limita a ejecutar instrucciones fijas; es capaz de aprender de la experiencia y mejorar con el tiempo. *Ejemplo:* un sistema de navegación como Google Maps aprende de las congestiones reportadas por otros usuarios y optimiza sus rutas en tiempo real.

- **Autonomía:** Capacidad de operar sin intervención humana continua. No significa que no necesite supervisión, sino que puede funcionar de forma independiente durante periodos prolongados. *Ejemplo:* un sistema de mantenimiento predictivo en una planta industrial monitoriza el estado de la maquinaria las 24 horas del día, los 7 días de la semana, y emite alertas solo cuando es necesario.

| Característica | Descripción |
|----------------|-------------|
| **Percepción** | Capacidad de captar información del entorno (sensores, datos, inputs). |
| **Procesamiento** | Capacidad de analizar, interpretar y transformar la información. |
| **Toma de decisiones** | Capacidad de seleccionar una acción entre varias posibles. |
| **Acción** | Capacidad de ejecutar la decisión seleccionada. |
| **Adaptabilidad** | Capacidad de aprender y mejorar con la experiencia. |
| **Autonomía** | Capacidad de operar sin intervención humana continua. |

### 1.3. Principios fundamentales

En este apartado vamos a explorar los principios que definen lo que hace que un sistema sea "inteligente" y no simplemente "automatizado". Comenzaremos distinguiendo entre inteligencia y automatización, dos conceptos que a menudo se confunden. A continuación, analizaremos los componentes básicos que conforman cualquier sistema inteligente. Por último, clasificaremos los sistemas según su nivel de inteligencia, desde los más simples hasta los teóricamente más avanzados.

#### 1.3.1. Inteligencia vs. Automatización

Es fundamental entender que **no todo lo automático es inteligente**. Esta distinción es la base de todo el módulo y afecta directamente a cómo diseñamos, implementamos y evaluamos los sistemas de IA en contextos reales.

**La automatización** es la ejecución de tareas preprogramadas, siguiendo reglas fijas que no cambian. Si las condiciones del entorno se desvían de lo previsto, el sistema automatizado no puede adaptarse; simplemente falla o produce un resultado incorrecto. Piensa en una línea de ensamblaje de una fábrica de coches: los robots realizan exactamente las mismas acciones una y otra vez, con una precisión milimétrica. Es increíblemente eficiente, pero totalmente incapaz de improvisar.

**La inteligencia**, en cambio, implica la capacidad de resolver problemas nuevos, adaptarse a contextos cambiantes y aprender de la experiencia. Un sistema inteligente no solo ejecuta reglas; puede generarlas, modificarlas o descartarlas según las circunstancias. *Ejemplo:* un sistema de diagnóstico médico inteligente no solo sigue un algoritmo predefinido; analiza el historial completo del paciente, considera factores contextuales y puede sugerir diagnósticos alternativos si los datos no encajan en el patrón habitual.

| Concepto | Definición | Ejemplo |
|----------|------------|---------|
| **Automatización** | Ejecución de tareas preprogramadas sin variación. | Una línea de ensamblaje robótica. |
| **Inteligencia** | Capacidad de resolver problemas nuevos, adaptarse y aprender. | Un sistema que aprende a reconocer objetos nunca vistos antes. |

> **Nota clave:** La automatización sigue reglas fijas. La inteligencia genera o modifica reglas según el contexto.

#### 1.3.2. Componentes básicos de un sistema inteligente

Todo sistema inteligente, independientemente de su complejidad o campo de aplicación, se construye sobre cuatro componentes fundamentales que trabajan en conjunto de forma secuencial. Comprender estos componentes es esencial para cualquier ingeniero o técnico que desee diseñar, implementar o mantener un sistema de IA.

El primer componente es la **percepción**, que actúa como los "sentidos" del sistema. A través de sensores, cámaras, micrófonos, APIs de datos o cualquier otro medio de captación de información, el sistema obtiene datos del entorno. Sin una buena percepción, el sistema no tiene materia prima con la que trabajar: "garbage in, garbage out". *Ejemplo:* un vehículo autónomo utiliza cámaras, lidar, radar y GPS para "ver" y "sentir" su entorno.

El segundo componente es el **procesamiento**, que constituye el "cerebro" del sistema. Aquí es donde los algoritmos, modelos de aprendizaje automático y bases de conocimiento transforman los datos crudos en información útil. Los datos se analizan, se buscan patrones, se comparan con conocimientos previos y se generan conclusiones. *Ejemplo:* un sistema de recomendación de películas procesa tus datos de visualización y los compara con las preferencias de millones de usuarios similares.

El tercer componente es la **toma de decisiones**, implementada a través de un motor de inferencia que evalúa las opciones disponibles y selecciona la acción más adecuada según los objetivos del sistema. *Ejemplo:* un sistema de gestión de energía en un edificio decide si encender o apagar el aire acondicionado basándose en la temperatura, la ocupación de las salas y el coste eléctrico actual.

El cuarto componente es la **acción**, mediante la cual el sistema ejecuta la decisión tomada actuando sobre el entorno. *Ejemplo:* el mismo sistema de gestión energética envía una señal a los termostatos para ajustar la temperatura.

```mermaid
flowchart LR
    A["👁️ PERCEPCIÓN<br/><br/>Sensores<br/>Cámaras<br/>Datos"]
    B["⚙️ PROCESAMIENTO<br/><br/>Algoritmos<br/>Modelos<br/>Bases de conocimiento"]
    C["🧠 TOMA DE DECISIÓN<br/><br/>Motor de inferencia"]
    D["🤖 ACCIÓN<br/><br/>Actuadores"]
    E["📚 APRENDIZAJE<br/><br/>Experiencia<br/>Retroalimentación<br/>Actualización del modelo"]

    A --> B
    B --> C
    C --> D
    D --> E
    E -.-> B
```

#### 1.3.3. Clasificación según su nivel de inteligencia

Los sistemas inteligentes no son todos iguales. Podemos clasificarlos según su nivel de inteligencia, desde los más simples (que solo reaccionan a estímulos) hasta los teóricamente más avanzados (que poseen autoconciencia). Esta clasificación nos ayuda a entender qué tipo de sistema necesitamos para cada aplicación y qué limitaciones podemos esperar.

| Nivel | Descripción | Ejemplo |
|-------|-------------|---------|
| **Reactivos** | Responden estímulos sin memoria ni estado interno. | Termostato básico. |
| **Con estado interno** | Mantienen información de estados anteriores. | Sistema de recomendación. |
| **Con modelo del mundo** | Construyen una representación del entorno. | Robot de navegación autónoma. |
| **Con consciencia (teórico)** | Tienen autoconciencia y comprensión profunda. | No existe actualmente. |

### 1.4. Paradigmas de la Inteligencia Artificial

En este apartado vamos a conocer los cuatro grandes paradigmas o enfoques históricos que han definido el desarrollo de la Inteligencia Artificial. Cada uno de ellos aporta una perspectiva diferente sobre cómo lograr la inteligencia en las máquinas: mediante reglas simbólicas, mediante redes neuronales inspiradas en el cerebro, mediante evolución artificial o mediante comportamientos emergentes. Comprender estos paradigmas es esencial para elegir la aproximación más adecuada en cada situación.

**El paradigma simbólico (también conocido como GOFAI — Good Old-Fashioned AI)** fue el primer enfoque serio de la IA. Se basa en representar el conocimiento del mundo mediante reglas lógicas y símbolos. Los sistemas simbólicos utilizan un motor de inferencia que aplica reglas "si-entonces" para llegar a conclusiones. Su gran ventaja es la explicabilidad: puedes seguir exactamente el razonamiento paso a paso. Sin embargo, son muy frágiles: si los datos incompletos o ambiguos no encajan en las reglas predefinidas, el sistema se bloquea. 
>*Ejemplo:* un sistema experto médico que diagnostica enfermedades aplicando reglas codificadas por médicos expertos.

**El paradigma conectivista (redes neuronales)** se inspira en el funcionamiento del cerebro humano. En lugar de programar reglas explícitamente, se entrena una red neuronal con grandes cantidades de datos para que aprenda por sí misma los patrones subyacentes. Su gran fortaleza es la robustez ante datos ruidosos o incompletos y su capacidad excepcional para el reconocimiento de patrones (imágenes, voz, texto). Su principal debilidad es la "caja negra": es difícil entender por qué toma una decisión concreta. 
>*Ejemplo:* una red neuronal convolucional que detecta tumores en radiografías con una precisión superior a la de los radiólogos humanos.

**El paradigma evolutivo** se inspira en el proceso de evolución biológica. Utiliza algoritmos genéticos que simulan la selección natural, el cruce y la mutación para explorar espacios de solución complejos. Es especialmente útil cuando no podemos definir una función de objetivo clara, pero podemos evaluar si una solución es "mejor" que otra. Su principal limitación es el elevado coste computacional. 
>*Ejemplo:* un algoritmo genético que optimiza el diseño de una pieza de avión para maximizar la resistencia y minimizar el peso.

**El paradigma comportamental** se basa en la idea de que comportamientos inteligentes complejos pueden emerger de la interacción de comportamientos simples. En lugar de programar un sistema completo de una vez, se diseñan comportamientos básicos que, al interactuar con el entorno y entre sí, generan comportamientos globales inteligentes. Es muy adaptativo y robusto, pero difícil de controlar y predecir. 
>*Ejemplo:* un enjambre de drones de búsqueda y rescate que, mediante reglas simples de evitación y búsqueda, cubren un área de forma eficiente sin un plan centralizado.

| Paradigma | Enfoque | Fortalezas | Debilidades |
|-----------|---------|------------|-------------|
| **Simbólico (GOFAI)** | Representación del conocimiento con reglas y lógica. | Explicabilidad, transparencia. | Fragilidad ante datos incompletos. |
| **Conectivista (Redes neuronales)** | Aprendizaje a partir de datos. | Robustez, reconocimiento de patrones. | Caja negra, necesidad de grandes datasets. |
| **Evolutivo** | Optimización mediante selección artificial. | Búsqueda en espacios complejos. | Coste computacional elevado. |
| **Comportamental** | Comportamientos emergentes a partir de interacciones simples. | Adaptabilidad, robustez. | Dificultad de control. |

## 3. Técnicas Básicas del Entorno de la IA

### 3.1. Visión general de técnicas

La Inteligencia Artificial no es una única técnica, sino un ecosistema de enfoques y algoritmos diseñados para resolver problemas de distinta naturaleza. Para entender cómo funcionan, podemos agrupar estas técnicas en grandes familias según su enfoque:

- **Técnicas basadas en el conocimiento (simbólicas):** Utilizan reglas lógicas y conocimiento experto codificado por humanos. Son muy explicables, pero rígidas.
- **Técnicas estadísticas y de aprendizaje automático (Machine Learning):** Los algoritmos "aprenden" de los datos en lugar de seguir reglas preprogramadas. Son flexibles y se adaptan a nuevos datos.
- **Técnicas inspiradas en la biología:** Redes neuronales (inspiradas en el cerebro), algoritmos genéticos (inspirados en la evolución) o lógica difusa (inspirada en el razonamiento humano aproximado).

A continuación se presentan las técnicas más relevantes para este módulo:

| Técnica | Descripción | Tipo de aprendizaje | Aplicación típica |
|---------|-------------|---------------------|-------------------|

### 3.2. Clasificación de técnicas por tipo de aprendizaje

No basta con elegir una técnica; es fundamental entender *cómo* aprende. En IA, el "aprendizaje" se refiere al proceso mediante el cual un sistema mejora su desempeño en una tarea mediante la experiencia. Dependiendo de la naturaleza de los datos con los que se entrena y del tipo de feedback que recibe, distinguimos tres grandes paradigmas de aprendizaje:

1. **Aprendizaje supervisado:** El modelo aprende a partir de datos etiquetados, donde se le proporciona la respuesta correcta para cada ejemplo. Es como un estudiante que resuelve ejercicios con soluciones al final del libro.
2. **Aprendizaje no supervisado:** El modelo trabaja con datos sin etiquetar y debe encontrar patrones o estructuras por sí mismo. Es como explorar un museo sin audioguía: descubres las conexiones tú mismo.
3. **Aprendizaje por refuerzo:** El agente aprende mediante la interacción directa con el entorno, recibiendo recompensas o penalizaciones por sus acciones. Es como aprender a andar en bicicleta: se aprende por ensayo y error, corrigiendo el equilibrio sobre la marcha.

#### 3.2.1. Aprendizaje supervisado

El modelo se entrena con datos etiquetados (entrada + respuesta correcta). Imagina que estás enseñando a un niño a reconocer animales: le muestras una foto de un perro y le dices "esto es un perro", una de un gato y le dices "esto es un gato". Con el tiempo, el niño aprende a asociar las características visuales (orejas, tamaño, color) con el nombre del animal. Así funciona el aprendizaje supervisado.

**Objetivo:** Aprender una función que mapee entradas a salidas conocidas.

El aprendizaje supervisado se divide en dos grandes categorías según el tipo de respuesta que se busca predecir:

- **Clasificación:** El objetivo es asignar una categoría o etiqueta discreta a una entrada. Es como hacer un examen de opción múltiple: el modelo debe elegir entre un conjunto finito de respuestas. *Ejemplo:* clasificar un correo como "spam" o "no spam", o diagnosticar si una radiografía muestra "tumor" o "sin tumor".
- **Regresión:** El objetivo es predecir un valor numérico continuo. Es como estimar una cantidad exacta. *Ejemplo:* predecir el precio de una vivienda en base a sus metros cuadrados, número de habitaciones y ubicación.

**Técnicas principales:**
- **Clasificación:** Asignar una categoría a una entrada (ej.: spam/no spam).
- **Regresión:** Predecir un valor continuo (ej.: precio de una casa).

| Algoritmo | Clasificación | Regresión |
|-----------|:----------:|:--------:|
| Regresión logística<sup>1</sup> | ✓ | |

><sup>2</sup> Una **SVM** es un algoritmo supervisado que busca encontrar una frontera de decisión que separe las clases maximizando el margen respecto a los datos más próximos, llamados vectores soporte.

><sup>3</sup> **Random Forest** y **Gradient Boosting** son familias de métodos, y pueden utilizarse tanto para clasificación como para regresión.

#### 3.2.2. Aprendizaje no supervisado

El modelo trabaja con datos sin etiquetar y busca patrones o estructuras por sí mismo. Imagina que recibes una caja llena de piezas de LEGO sin instrucciones ni categorías. No sabes qué piezas van con qué; tu tarea es organizarlas. Agruparías las piezas por color, por tamaño o por forma. Eso es exactamente lo que hace el aprendizaje no supervisado: toma datos sin clasificar y descubre por sí mismo las estructuras subyacentes.

**Objetivo:** Descubrir patrones ocultos en los datos.

A diferencia del supervisado, aquí no hay un "profesor" que diga si la respuesta es correcta o incorrecta. El sistema explora los datos y encuentra regularidades, agrupaciones o relaciones que un humano podría pasar por alto.

#### 3.2.3. Aprendizaje por refuerzo

Un agente aprende a tomar decisiones mediante interacción con el entorno, recibiendo recompensas o penalizaciones. Piensa en cómo un perro aprende un truco: le enseñas un comando, el perro hace una acción, si lo hace bien recibe una premiación (un premio o una caricia), si lo hace mal no recibe nada. Con el tiempo, el perro aprende a asociar ciertas acciones con recompensas.

Este tipo de aprendizaje no se basa en ejemplos etiquetados, sino en la **experiencia directa**. El agente prueba acciones, observa las consecuencias y ajusta su comportamiento para maximizar la recompensa acumulada a largo plazo.

**Componentes:**
- **Agente:** El sistema que toma decisiones.
- **Entorno:** El mundo con el que interactúa.
- **Estado (s):** La situación actual del agente.
- **Acción (a):** Lo que el agente puede hacer.
- **Recompensa (r):** Feedback numérico de la acción.
- **Política (π):** Estrategia del agente para elegir acciones.

#### 3.5.1. Métricas de evaluación

Evaluar un modelo de IA no es tan simple como decir "acierta el 90% de las veces". Dependiendo del problema, una métrica puede ser engañosa. Por ejemplo, si intentamos predecir una enfermedad rara que afecta al 1% de la población, un modelo que siempre diga "no tiene la enfermedad" tendría un 99% de acierto, pero sería inútil porque no detecta a ningún enfermo. Por eso, necesitamos métricas específicas para cada situación.

A continuación se explican las métricas más importantes:

- **Accuracy (Precisión global):** Es el porcentaje total de predicciones correctas. Solo es útil cuando las clases están equilibradas (misma cantidad de ejemplos de cada categoría).
- **Precision (Precisión positiva):** De todas las veces que el modelo predijo "positivo", ¿cuántas eran realmente positivas? Es crucial cuando el coste de un falso positivo es alto (ej.: diagnosticar erróneamente una enfermedad a un sano).
- **Recall (Sensibilidad o Recall):** De todas las veces que era realmente "positivo", ¿cuántas logró detectar el modelo? Es crucial cuando el coste de un falso negativo es grave (ej.: no detectar un tumor).
- **F1-Score:** Es el equilibrio entre Precision y Recall. Cuando necesitas un balance entre ambos, el F1-Score te da un único número que refleja el rendimiento global.
- **ROC-AUC:** Mide la capacidad del modelo para distinguir entre clases en todos los umbrales posibles. Un valor de 1.0 es perfecto; 0.5 es aleatorio. Se representa:

    

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

# Semilla para obtener siempre los mismos resultados
np.random.seed(42)

# 100 valores reales: 50 negativos y 50 positivos
y_real = np.array([0] * 50 + [1] * 50)

# Probabilidades predichas por el modelo
# Los positivos tienden a recibir probabilidades más altas
y_pred_negativos = np.random.normal(0.30, 0.20, 50)
y_pred_positivos = np.random.normal(0.70, 0.20, 50)

y_pred = np.concatenate([
    y_pred_negativos,
    y_pred_positivos
])

# Limitar las probabilidades al intervalo [0, 1]
y_pred = np.clip(y_pred, 0, 1)

# Calcular la curva ROC
fpr, tpr, thresholds = roc_curve(y_real, y_pred)

# Calcular AUC
roc_auc = auc(fpr, tpr)

# Dibujar la curva
plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    linewidth=2,
    label=f"Curva ROC (AUC = {roc_auc:.2f})"
)

# Clasificador aleatorio
plt.plot(
    [0, 1],
    [0, 1],
    "--",
    linewidth=1.5,
    label="Clasificador aleatorio (AUC = 0.50)"
)

# Área bajo la curva
plt.fill_between(
    fpr,
    tpr,
    alpha=0.2
)

plt.xlim(0, 1)
plt.ylim(0, 1.05)

plt.xlabel("Tasa de falsos positivos (FPR)")
plt.ylabel("Tasa de verdaderos positivos (TPR)")
plt.title("Curva ROC")

plt.legend(loc="lower right")
plt.grid(True, alpha=0.3)

plt.show()
````

- **MSE (Error cuadrático medio) y MAE (Error absoluto medio):** Se usan en regresión. El MSE penaliza más los errores grandes (por cuadrar los errores), mientras que el MAE da una idea más directa del error medio en las mismas unidades que la variable objetivo.

| Métrica | Uso | Fórmula / concepto |
|---|---|---|
| **Accuracy** | Clasificación, especialmente con clases equilibradas. | $$\frac{TP + TN}{TP + TN + FP + FN}$$ |
| **Precision** | Proporción de predicciones positivas que son realmente positivas. | $$\frac{TP}{TP + FP}$$ |
| **Recall (Sensibilidad)** | Proporción de positivos reales que el modelo consigue detectar. | $$\frac{TP}{TP + FN}$$ |
| **F1-Score** | Equilibrio entre Precision y Recall. | $$2\frac{Precision \cdot Recall}{Precision + Recall}$$ |
| **ROC-AUC** | Capacidad del modelo para discriminar entre clases. | Área bajo la curva ROC. |
| **MSE (Error cuadrático medio)** | Regresión. | $$\frac{\sum (y_i-\hat{y_i})^2}{n}$$ |
| **MAE (Error absoluto medio)** | Regresión. | $$\frac{\sum \lvert y_i-\hat{y}_i\rvert}{n}$$ |

Leyenda: 
- $$y_i$$ → valor real.
- $$\hat{y}_i$$ → valor predicho por el modelo.
- $$n$$ → número de ejemplos.
- $$TP$$ → verdaderos positivos.
- $$TN$$ → verdaderos negativos.
- $$FP$$ → falsos positivos.
- $$FN$$ → falsos negativos.


#### 3.5.2. Sobreajuste (Overfitting) e Underfitting

Uno de los mayores desafíos al construir un modelo de IA es encontrar el "punto justo" de complejidad. Si el modelo es demasiado simple, no aprende nada útil. Si es demasiado complejo, aprende tanto los patrones reales como el "ruido" (las irregularidades aleatorias) de los datos de entrenamiento. A estos dos extremos se les conoce como *Underfitting* y *Overfitting*.

Imagina que un estudiante se prepara para un examen:
- **Underfitting (Subajuste):** El estudiante no estudia lo suficiente. No entiende los conceptos básicos y reprueba el examen, pero también reprueba cualquier examen similar. Es un modelo demasiado simple que no captura la complejidad de los datos.
- **Overfitting (Sobreajuste):** El estudiante se memoriza las preguntas y respuestas exactas del examen de práctica, pero no entiende los conceptos. Cuando le hacen preguntas ligeramente diferentes en el examen real, reprueba. Es un modelo que memoriza los datos de entrenamiento pero no sabe generalizar a situaciones nuevas.

El objetivo es encontrar el equilibrio: un modelo lo suficientemente complejo para capturar los patrones reales, pero no tanto como para memorizar el ruido.

| Fenómeno | Descripción | Causa | Solución |
|----------|-------------|-------|----------|
| **Underfitting** | El modelo no captura los patrones de los datos. | Modelo demasiado simple. | Aumentar complejidad, más características. |
| **Sobreajuste (Overfitting)** | El modelo memoriza los datos de entrenamiento pero no generaliza. | Modelo demasiado complejo, pocos datos. | Regularización, más datos, Dropout, Early stopping. |

#### 3.5.3. División de datos (Train/Validation/Test)

| Conjunto | Propósito | Porcentaje típico |
|----------|-----------|-------------------|
| **Entrenamiento (Train)** | Ajustar los parámetros del modelo. | 60-80% |
| **Validación (Val)** | Ajustar hiperparámetros y selección de modelo. | 10-20% |
| **Test** | Evaluar el rendimiento final sin sesgo. | 10-20% |

> **Nota:** En datasets pequeños se usa **Cross-Validation** (K-Fold) para maximizar el uso de datos.

- Reducción de pérdidas por robos.
- Optimización del espacio y flujo de trabajo.
- Experiencia personalizada.

> **Ejemplo práctico:** Una fábrica de componentes electrónicos utiliza visión por computadora para inspeccionar miles de piezas por hora. Cada pieza pasa ante una cámara de alta resolución, y un modelo de deep learning (entrenado con miles de imágenes de piezas defectuosas) analiza en tiempo real si hay micro-grietas, deformaciones o soldaduras incorrectas. El sistema detecta defectos que un inspector humano pasaría por alto por fatiga, y lo hace a una velocidad 10 veces superior. El resultado: la tasa de defectos que llegan al cliente se reduce de un 2% a menos del 0.1%, y se ahorra el coste de los devoluciones y reclamaciones.

#### 4.2.6. Interacción mediante interfaces conversacionales avanzadas

**Descripción:** Sistemas que mantienen conversaciones naturales y contextuales con los usuarios.

**Evolución:**

```mermaid

    A["Interacción<br/>tradicional"]
    B["Menús<br/>interactivos"]
    C["Chatbots<br/>básicos"]
    D["Asistentes<br/>conversacionales"]
    E["Modelos de lenguaje<br/>generativos (LLM)"]

    A --> B --> C --> D --> E
```

**Características de los LLM (Large Language Models):**
- Entrenados con enormes cantidades de texto.
- Capacidad de generar texto coherente y contextual.
- Adaptabilidad a múltiples tareas sin reentrenamiento.
- Ejemplos: GPT, Claude, LLaMA, Gemini.

**Beneficios para la eficiencia operativa:**
- Interacción natural con clientes y empleados.
- Generación automática de contenido y documentos.
- Asistencia en toma de decisiones complejas.
- Reducción de barreras de acceso a la tecnología.

> **Ejemplo práctico:** Un despacho de abogados implementa un modelo de lenguaje (LLM) fine-tuneado con jurisprudencia y contratos para revisar documentos legales. El abogado carga un contrato de arrendamiento de 50 páginas, y el sistema lo analiza en minutos, identificando cláusulas riesgosas, comparándolas con la legislación vigente y sugiriendo modificaciones. Lo que antes requería 4-6 horas de revisión manual, ahora se completa en menos de 10 minutos, con una cobertura del 95% de los casos estándar. El abogado solo revisa los casos excepcionales que el sistema marca como de alto riesgo. El resultado: el despacho puede atender 5 veces más clientes sin contratar más personal.

### 4.3. Análisis de casos: Eficiencia operativa

Para cerrar esta sección, vamos a ver tres casos reales que muestran el antes y después de implementar IA en procesos empresariales clave. En cada caso, se compara el estado actual con y sin inteligencia artificial, destacando las mejoras cuantificables en eficiencia, coste y calidad. Los casos son representativos de sectores muy distintos (banca, comercio electrónico y manufactura), lo que demuestra que los principios de la IA aplicable a negocios son transversales.

#### Caso 1: Banco → Detección de fraude

| Tiempo de respuesta | Horas/días | Milisegundos |
| Ahorro estimado | — | 30-50% en costes de fraude |

#### Caso 2: E-commerce → Recomendaciones

| Ticket medio | Base | +15-30% |

#### Caso 3: Manufactura → Control de calidad

### 4.4. Tendencias emergentes

La inteligencia artificial no es un producto estático; es un campo en constante evolución. Cada año aparecen nuevas capacidades, nuevas formas de aplicar la tecnología y nuevas preguntas éticas que resolver. En este apartado vamos a repasar las cinco tendencias más importantes que están definiendo el futuro de la IA en los negocios. Cada una de ellas responde a una necesidad concreta: hacer la IA más útil, más segura, más accesible o más justa.

#### IA generativa

La **IA generativa** es, sin duda, la tendencia más mediática de los últimos años. Se refiere a modelos capaces de crear contenido nuevo —texto, imágenes, vídeo, música, código— a partir de lo que han aprendido de grandes volúmenes de datos. No se limita a clasificar o predecir; *crea* algo que no existía antes. Piensa en un modelo al que le pides: "Escribe un informe de ventas trimestral con gráficos, un resumen ejecutivo y recomendaciones estratégicas". Un sistema generativo puede producir todo eso desde cero, adaptando el tono, el formato y el contenido a tu solicitud.

**Impacto en los negocios:** Las empresas están usando la IA generativa para acelerar procesos que antes requerían horas o días de trabajo humano: redacción de informes, diseño de maquetas, generación de código, creación de material de formación. No reemplaza al profesional; lo multiplica.

#### IA en el edge (Edge AI)

La **IA en el edge** consiste en ejecutar modelos de inteligencia artificial directamente en el dispositivo del usuario —un teléfono, una cámara de seguridad, un sensor industrial— en lugar de enviar los datos a un servidor central en la nube. Imagina que tu teléfono puede reconocer tu rostro para desbloquearse sin necesidad de conectarse a ningún servidor: eso es IA en el edge.

**Impacto en los negocios:** Al procesar los datos localmente, se reduce la latencia (no hay que esperar a que los datos viajen a un servidor), se mejora la privacidad (los datos sensibles no salen del dispositivo) y se reduce el coste de ancho de banda. Es especialmente relevante en sectores como la salud, la industria o la seguridad.

#### IA explicable (XAI — eXplainable AI)

Uno de los mayores problemas de los modelos de IA más potentes —especialmente las redes neuronales profundas— es que son "cajas negras": sabemos qué entrada reciben y qué salida producen, pero no sabemos *cómo* llegan a esa conclusión. La **IA explicable** busca abrir esa caja negra, proporcionando razones comprensibles para las decisiones del modelo. Si un banco rechaza una solicitud de préstamo, el cliente tiene derecho a saber por qué. La XAI proporciona esa explicación.

**Impacto en los negocios:** Permite cumplir con normativas cada vez más exigentes (como el Reglamento General de Protección de Datos o la Ley de IA de la UE), genera confianza en los usuarios y facilita la auditoría de los sistemas. Sin explicabilidad, muchos sectores simplemente no pueden usar IA.

#### IA ética y responsable

La **IA ética y responsable** no es una tecnología en sí misma, sino un marco de principios y prácticas que guían el desarrollo y despliegue de sistemas de IA. Se centra en evitar sesgos discriminatorios, proteger la privacidad, garantizar la transparencia y asegurar que los sistemas de IA beneficien a toda la sociedad, no solo a un grupo reducido. Un sistema de contratación que descarta automáticamente a candidatas mujeres por sesgo en los datos de entrenamiento no es ético; un sistema de IA responsable se diseña para detectar y corregir esos sesgos.

**Impacto en los negocios:** Evita sanciones legales, protege la reputación de la marca y construye confianza con clientes y empleados. Cada vez más, los consumidores y reguladores exigen que la IA se use de forma responsable.

#### IA como servicio (AIaaS — AI as a Service)

La **IA como servicio** permite a cualquier empresa acceder a capacidades de inteligencia artificial mediante una API o suscripción, sin necesidad de construir modelos propios ni contratar equipos de científicos de datos. Es el equivalente a alquilar un modelo de IA en lugar de construirlo desde cero. Piensa en ello como un servicio de streaming: en lugar de fabricar tus propias películas, pagas una suscripción para acceder a un catálogo enorme. Con AIaaS, en lugar de entrenar un modelo de procesamiento de lenguaje natural desde cero (lo que costaría millones), pagas por usar uno que ya existe.

**Impacto en los negocios:** Democratiza el acceso a la IA, reduce drásticamente los costes de implementación y permite a empresas pequeñas y medianas competir con grandes corporaciones. Es la barrera de entrada más baja que hemos visto jamás.

| Tendencias | Descripción | Impacto en eficiencia |
|------------|-------------|----------------------|
| **IA generativa** | Creación de contenido (texto, imagen, vídeo, código). | Aceleración de procesos creativos y analíticos. |
| **IA en el edge** | Procesamiento de IA en dispositivos locales. | Menor latencia, mayor privacidad. |
| **IA explicable (XAI)** | Modelos que explican sus decisiones. | Mayor confianza, cumplimiento normativo. |
| **IA ética y responsable** | Diseño de sistemas justos y transparentes. | Reducción de riesgos legales y reputacionales. |
| **IA como servicio (AIaaS)** | Acceso a capacidades de IA mediante API. | Democratización, reducción de costes de implementación. |
