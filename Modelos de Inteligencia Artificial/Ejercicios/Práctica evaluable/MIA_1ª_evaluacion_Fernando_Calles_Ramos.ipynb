## Ejercicio 2: Búsqueda en Anchura (BFS)

**Problema:** Dado un grafo que representa ciudades conectadas por carreteras, encontrar la ruta más corta (en número de saltos) desde una ciudad origen hasta una ciudad destino usando BFS.

```python
from collections import deque


def bfs(grafo, inicio, objetivo):
    cola = deque([[inicio]])
    visitados = {inicio}

    while cola:
        camino = cola.popleft()
        nodo = camino[-1]

        if nodo == objetivo:
            return camino

        for vecino in grafo[nodo]:
            if vecino not in visitados:
                visitados.add(vecino)
                cola.append(camino + [vecino])

    return None


def encontrar_rutas_todas(grafo, inicio, objetivo):
    rutas = []
    cola = deque([[inicio]])

    while cola:
        camino = cola.popleft()
        nodo = camino[-1]

        if nodo == objetivo:
            rutas.append(camino)
            continue

        for vecino in grafo[nodo]:
            if vecino not in camino:
                cola.append(camino + [vecino])

    return rutas


grafo_ciudades = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "E"],
    "D": ["B", "F", "G"],
    "E": ["B", "C", "G"],
    "F": ["D"],
    "G": ["D", "E"]
}


print("Ruta más corta:")
ruta = bfs(grafo_ciudades, "A", "F")

if ruta is not None:
    print(" - ".join(ruta))
else:
    print("No existe ninguna ruta.")


print("\nTodas las rutas:")
for ruta in encontrar_rutas_todas(grafo_ciudades, "A", "F"):
    print(" - ".join(ruta))
```

**Pregunta de reflexión:** ¿Por qué BFS garantiza encontrar siempre la ruta más corta, pero no todas las rutas? ¿En qué casos sería preferible DFS a BFS?

-	BFS explora el grafo por niveles, primero los nodos situados a un salto del origen, luego los situados a dos y así sucesivamente. La primera ruta que encuentra hasta el destino tiene será la que tenga el menor número de saltos.
-	DFS es preferible cuando solo necesitamos encontrar un camino valido, sin comprobar que sea el más corto. También puede ser útil en sistemas con poca memoria, ya que en búsquedas con muchas ramificaciones (un grafo más grande) no va a profundizar tanto y no necesita acumular rutas, solo devolver la primera que encuentre.

## Ejercicio 3: Búsqueda en Profundidad (DFS)

**Problema:** Dado un laberinto representado como un grafo, encontrar un camino desde la entrada hasta la salida usando DFS.

```python
def dfs(grafo, inicio, objetivo):
    """
    Búsqueda en profundidad (DFS) iterativa que devuelve un camino
    desde 'inicio' hasta 'objetivo'.
    
    Args:
        grafo: dict — grafo de adyacencia.
        inicio: str — nodo de partida.
        objetivo: str — nodo destino.
    
    Returns:
        list — camino encontrado, o None si no existe.
    """
    pila = [(inicio, [inicio])] # (nodo_actual, camino)
    visitados = {inicio}

    while pila:
        nodo, camino = pila.pop()  # Sacar último añadido (LIFO)

        if nodo == objetivo:
            return camino

        for vecino in grafo[nodo]:
            if vecino not in visitados:
                visitados.add(vecino)
                pila.append((vecino, camino + [vecino]))

    return None

def dfs_con_rastro(grafo, inicio, objetivo):
    pila = [(inicio, [inicio])]
    visitados = {inicio}
    rastro = []

    while pila:
        nodo, camino = pila.pop()
        rastro.append(nodo)
        print("Explora: ",{nodo})

        if nodo == objetivo:
            print("Rastro: ", " - ".join(rastro))
            return camino

        for vecino in grafo[nodo]:
            if vecino not in visitados:
                visitados.add(vecino)
                pila.append((vecino, camino + [vecino]))

    print("Rastro: ", " ->".join(rastro))
    return None

almacen = {
    "ENTRADA": ["A", "B"],
    "A": ["ENTRADA", "B", "C"],
    "B": ["ENTRADA", "A", "E"],
    "C": ["A", "D"],
    "D": ["C", "F"],
    "E": ["B", "F"],
    "F": ["D", "E", "SALIDA"],
    "SALIDA": ["F"]
}
print("Camino con DFS:")
camino = dfs(almacen, "ENTRADA", "SALIDA")

if camino is not None:
    print(" - ".join(camino))
else:
    print("No existe un camino.")

camino = dfs_con_rastro(almacen, "ENTRADA", "SALIDA")
print(f"Camino: {" -> ".join(camino)}")
print("Camino encontrado:", " - ".join(camino))
```

**Pregunta de reflexión:** Si el almacén tuviera un ciclo largo sin salida (por ejemplo, un pasillo circular), ¿qué problema tendría DFS? ¿Cómo se podría mitigar?

-	DFS podría recorrer los mismos nodos infinitamente si no controlase los visitados, impidiendo explorar otros caminos.
-	Por ello, se usa un conjunto de visitados, que impide añadir a la pila nodos por los que ya hemos pasado.
