```python
from collections import deque

def encontrar_rutas_todas(grafo, inicio, objetivo):
    cola = deque([[inicio]])      # Cola de caminos pendientes
    rutas = []   

    while cola:
        camino = cola.popleft()        # Sacar el camino más antiguo
        print("v",camino)
        print("n",camino[-1])
        nodo = camino[-1]              # Último nodo del camino actual

        if nodo == objetivo:           # ¿Hemos llegado al objetivo?
            rutas.append(camino)
            continue

        print("G", grafo[nodo])
        for vecino in grafo[nodo]:     # Explorar vecinos
            if vecino not in camino:
                cola.append(camino + [vecino])
                print("cola añadida", camino + [vecino])

    return rutas                        # No hay camino

grafo_ciudades = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "E"],
    "D": ["B", "F", "G"],
    "E": ["B", "C", "G"],
    "F": ["D"],
    "G": ["D", "E"],
}

ruta = encontrar_rutas_todas(grafo_ciudades, "A", "F")
print(ruta)
```
