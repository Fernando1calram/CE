```python
def dfs_con_rastro(grafo, inicio, objetivo):
    pila = [(inicio, [inicio])] # (nodo_actual, camino)
    visitados = {inicio}
    rastro = []

    while pila:
        print("Pila", pila)
        print("Visitados", visitados)
        nodo, camino = pila.pop()  # Sacar último añadido (LIFO)
        rastro.append(nodo)

        print("Explorando: ", nodo)

        if nodo == objetivo:
            print("Rastro de exploración: ", " -> ".join(rastro))
            return camino

        for vecino in grafo[nodo]:
            if vecino not in visitados:
                visitados.add(vecino)
                pila.append((vecino, camino + [vecino]))

    print("Rastro de exploración: ", " -> ".join(rastro))

    return None

almacen = {
    "ENTRADA": ["A","B"],
    "A": ["ENTRADA","B","C"],
    "B": ["ENTRADA","A","E"],
    "C": ["A", "D"],
    "D": ["C","F"],
    "E": ["B","F"],
    "F": ["SALIDA","D","E"],
    "SALIDA": ["F"]
}

camino = dfs_con_rastro(almacen, "ENTRADA", "SALIDA")
print(f"Camino: {" -> ".join(camino)}")
```
