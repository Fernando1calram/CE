## Ejercicio 2: Búsqueda en Anchura (BFS)

**Problema:** Dado un grafo que representa ciudades conectadas por carreteras, encontrar la ruta más corta (en número de saltos) desde una ciudad origen hasta una ciudad destino usando BFS.

```python
from collections import deque

def encontrar_rutas_todas(grafo, inicio, objetivo):
    rutas = []
    cola = deque([[inicio]])

    while cola: 
        camino = cola.popleft()
        print("camino", camino)
        nodo = camino[-1]

        if nodo == objetivo:
            rutas.append(camino)
            continue

        for vecino in grafo[nodo]:
            if vecino not in camino:
                cola.append(camino + [vecino])
    return rutas

grafo_ciudades = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "E"],
    "D": ["B", "F", "G"],
    "E": ["B", "C", "G"],
    "F": ["D"],
    "G": ["D","E"]
}

print("Todas las rutas:")
for ruta in encontrar_rutas_todas(grafo_ciudades, "A", "G"):
    print(" - ".join(ruta))
```

## Ejercicio 3: Búsqueda en Profundidad (DFS)

**Problema:** Dado un laberinto representado como un grafo, encontrar un camino desde la entrada hasta la salida usando DFS.

```python
def dfs_con_rastro(grafo, inicio, objetivo):
    pila = [(inicio, [inicio])]
    visitados = {inicio}
    rastro = []

    while pila:
        nodo, camino = pila.pop()
        rastro.append(nodo)
        print("Explora: ",{nodo})

        if nodo == objetivo:
            print("Rastro: ", " - ".join(rastro))
            return camino

        for vecino in grafo[nodo]:
            visitados.add(vecino)
            pila.append((vecino, camino + [vecino]))

    print("Rastro: ", " ->".join(rastro))
    return None

laberinto = {
    "ENTRADA": ["A", "B"],
    "A": ["ENTRADA", "C"],
    "B": ["ENTRADA", "A", "E"],
    "C": ["A", "D"],
    "D": ["C", "F"],
    "E": ["B", "F"],
    "F": ["D", "D", "SALIDA"],
    "SALIDA": ["F"]
}

camino = dfs_con_rastro(laberinto, "ENTRADA", "SALIDA")
print(f"Camino: {" -> ".join(camino)}")
print("Camino encontrado:", " - ".join(camino))
```
