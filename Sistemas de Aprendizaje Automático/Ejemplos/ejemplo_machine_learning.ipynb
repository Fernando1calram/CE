```python
import pandas as pd
```

```python
melb_df = pd.read_csv('melb_data.csv')

# Muestra el csv leído
#melb_df

# Dice lo que tiene cada columna
melb_df.describe()
```

```python
# Quitamos cualquier fila que tenga un missing value
melb_df = melb_df.dropna(axis = 0)
melb_df

melb_df.describe()
```

```python
# La variable a predecir va a ser "Price". Variable y
# Las variables numéricas para entrenar el modelo y realizar la predicción del precio, las que considero más idóneas. Las x son los predictores
y = melb_df['Price']
x = melb_df[['Rooms','Bathroom','Landsize','Lattitude','Longtitude']]
```

```python
# Ejemplo con un DecisionTreeRegresor. Para variables númericas. Le vamos a pasar todas las filas cribadas
from sklearn.tree import DecisionTreeRegressor
melbourne_model = DecisionTreeRegressor()
melbourne_model.fit(x,y)
```

```python
# Predicciones para las 5 primeras casas
print("Prediciones para las 5 primeras casas: ")
print(x.head())
print("El precio real es: ")
print(y.head())
print("Las predicciones son: ")
print(melbourne_model.predict(x.head()))
```

```python
# MAE (Mean Absolute Error) Calcula el error que comete la predicción por cada y los suma todos
# Se hace en términos absolutos. Listamos haciendo las predicciones sobre las casas con las que entrenamos el modelo
from sklearn.metrics import mean_absolute_error
predicted_home_prices = melbourne_model.predict(x)
mean_absolute_error(y, predicted_home_prices)
```

```python
# Si tengo 10.000 filas con datos, la técnica en MD es dividir en 5.000 para entrenar y 5.000 para predecir.
# x para entrenar y x para validar
from sklearn.model_selection import train_test_split
train_x, val_x, train_y, val_y = train_test_split(x,y)

melbourne_model = DecisionTreeRegressor()
melbourne_model.fit(train_x, train_y)

val_predictions = melbourne_model.predict(val_x)
print(mean_absolute_error(val_y, val_predictions))
```

```python
# Vamos a probar con 5, 50, 500, 5.000 a ver con cuantas filas tenemos la mejor predicción.
# El mejor es con 500
def get_mae(max_leaf_nodes, train_x, val_x, train_y, val_y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes)
    model.fit(train_x, train_y)
    preds_val = model.predict(val_x)
    mae = mean_absolute_error(val_y, preds_val)
    return(mae)

for max_leaf_nodes in [5, 50, 500, 5000]:
    my_mae = get_mae(max_leaf_nodes, train_x, val_x, train_y, val_y)
    print("Max leaf nodes: %d  \t\t Mean Absolute Error:  %d" %(max_leaf_nodes, my_mae))
```
