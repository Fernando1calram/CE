<h1><b>Práctica evaluable</b><br>Sistema inteligente de monitorización y diagnóstico industrial</h1>
</div>

## 1. Contexto

Una empresa industrial quiere desarrollar un sistema de monitorización que permita supervisar el estado de sus máquinas mediante los datos recogidos por diferentes sensores.

Cada máquina dispone de sensores que registran periódicamente las siguientes variables:

- **Temperatura:** temperatura de funcionamiento del motor (°C).
- **Vibración:** intensidad de las vibraciones (mm/s).
- **Consumo eléctrico:** potencia consumida (kW).
- **Presión:** presión del circuito hidráulico (bar).

A partir de estas mediciones, la empresa quiere disponer de una herramienta capaz de analizar el comportamiento de las máquinas, identificar situaciones anómalas y estimar el tipo de incidencia que puede estar produciéndose.

Se solicita desarrollar una aplicación en Python que permita procesar los datos, realizar un análisis estadístico y construir un sistema básico de clasificación utilizando NumPy.

Este trabajo práctico comprende una pequeña parte de investigación sobre algunos conceptos solicitados, con el fin de que el alumno/a profundice en la comprensión de diferentes conceptos clave.

## 2. Objetivos

La aplicación deberá permitir:

1. Cargar y gestionar conjuntos de datos mediante arrays de NumPy.
2. Identificar y tratar valores ausentes, incorrectos y anómalos.
3. Realizar análisis estadísticos de las variables registradas.
4. Aplicar operaciones vectorizadas para transformar y normalizar los datos.
5. Implementar un sistema básico de clasificación sin utilizar bibliotecas de aprendizaje automático.
6. Evaluar el rendimiento del sistema mediante diferentes métricas.
7. Presentar los resultados de manera estructurada.

## 3. Datos de partida

Se trabajará con un conjunto de datos denominado `sensores.csv`.

Cada registro representa una medición realizada sobre una máquina.

El archivo contendrá las siguientes columnas:

| Columna     | Descripción                    |
| ----------- | ------------------------------ |
| id          | Identificador del registro     |
| temperatura | Temperatura en °C              |
| vibracion   | Vibración en mm/s              |
| consumo     | Consumo eléctrico en kW        |
| presion     | Presión en bar                 |
| estado      | Diagnóstico real de la máquina |

La variable `estado` podrá tomar los siguientes valores:

- `0`: Funcionamiento normal.
- `1`: Desgaste mecánico.
- `2`: Sobrecalentamiento.

El conjunto de datos contendrá registros correspondientes a las tres categorías, así como algunos valores ausentes, mediciones incorrectas y observaciones anómalas.

Se proporcionará un archivo de datos inicial para realizar la práctica.

## 4. Desarrollo de la práctica

### Apartado 1. Carga y exploración de los datos

Implementar las funciones necesarias para cargar el archivo CSV utilizando Python y NumPy.

La aplicación deberá:

- Obtener las dimensiones del conjunto de datos.
- Identificar las variables disponibles.
- Mostrar una muestra de los registros.
- Comprobar los tipos de datos.
- Identificar los valores ausentes.
- Calcular el número de registros correspondientes a cada categoría.

Se deberá comprobar que los datos se han cargado correctamente antes de continuar.

### Apartado 2. Limpieza y preparación de los datos

Desarrollar un procedimiento de limpieza que permita evaluar la calidad del conjunto de datos.

La aplicación deberá:

1. Identificar los registros que contienen valores ausentes y establecer un procedimiento para tratarlos.
2. Detectar mediciones que se encuentren fuera de los rangos físicos admisibles de los sensores.
3. Identificar posibles valores atípicos mediante técnicas estadísticas.
4. Diferenciar entre errores de medición y observaciones extremas que puedan representar situaciones reales de funcionamiento anómalo.
5. Separar los registros válidos de aquellos que deban descartarse o corregirse.

Para realizar esta tarea se proporcionan los siguientes rangos físicos admisibles:

| Variable | Valor mínimo | Valor máximo |
|---|---:|---:|
| Temperatura (°C) | 0 | 150 |
| Vibración (mm/s) | 0 | 15 |
| Consumo eléctrico (kW) | 0 | 65 |
| Presión (bar) | 0 | 12 |

**Consideraciones:**

- No se deben eliminar automáticamente todos los valores extremos.
- La identificación de un valor atípico mediante un criterio estadístico no implica necesariamente que sea un error.
- Se deberán justificar las decisiones tomadas durante el proceso de limpieza.
- Se deberá explicar cómo afecta el tratamiento de los datos a las fases posteriores de análisis y clasificación.

### Apartado 3. Análisis estadístico

Realizar un análisis exploratorio de las variables numéricas.

Para cada variable se deberán calcular:

- Media.
- Mediana.
- Desviación estándar.
- Valor mínimo y máximo.
- Primer y tercer cuartil.
- Rango intercuartílico.

Además, se deberá realizar un análisis comparativo de estas variables para cada una de las tres categorías de funcionamiento.

El objetivo es identificar qué características permiten diferenciar los distintos estados de las máquinas.

### Apartado 4. Transformación y normalización

Preparar los datos para construir un sistema de clasificación.

Se deberá:

- Separar las variables predictoras de la variable objetivo.
- Dividir el conjunto de datos en entrenamiento (80 %) y prueba (20 %).
- Mantener una distribución equilibrada de las categorías en ambas particiones.
- Normalizar las variables utilizando la media y desviación estándar del conjunto de entrenamiento.
- Aplicar exactamente la misma transformación al conjunto de prueba.

Todas estas operaciones deberán implementarse utilizando NumPy.

### Apartado 5. Desarrollo del sistema de clasificación

Implementar desde cero un clasificador basado en el método del **centroide más cercano**.

El funcionamiento será el siguiente:

1. Calcular el centroide de cada categoría utilizando los datos de entrenamiento.
2. Para cada observación del conjunto de prueba, calcular su distancia euclídea respecto a los tres centroides.
3. Asignar a la observación la categoría cuyo centroide se encuentre a menor distancia.
4. Almacenar las predicciones obtenidas.

No se permite utilizar Scikit-learn ni otras bibliotecas que implementen algoritmos de clasificación.

Se valorará especialmente el uso de operaciones vectorizadas de NumPy frente a la utilización innecesaria de bucles.

### Apartado 6. Evaluación del clasificador

Evaluar el rendimiento del sistema desarrollado.

Se deberán calcular:

- Matriz de confusión.
- Accuracy.
- Precision por categoría.
- Recall por categoría.
- F1-score por categoría.

Las métricas deberán implementarse manualmente utilizando operaciones de NumPy.

A partir de los resultados, responder:

1. ¿Qué categoría presenta mejores resultados?
2. ¿Qué tipo de incidencias se confunden con mayor frecuencia?
3. ¿Existe alguna categoría que el sistema tenga dificultades para identificar?
4. ¿Consideras que el clasificador sería adecuado para utilizarse en un entorno industrial real? Justifica la respuesta.

### Apartado 7. Análisis y conclusiones

Elaborar un informe final que recoja:

- Descripción del conjunto de datos.
- Problemas encontrados durante la limpieza.
- Principales resultados del análisis estadístico.
- Funcionamiento del clasificador.
- Resultados de la evaluación.
- Limitaciones del sistema.
- Posibles mejoras.

Se deberá justificar si el sistema desarrollado cumple los objetivos iniciales y qué aspectos deberían mejorarse antes de utilizarlo en un entorno real.

## 5. Requisitos técnicos

La solución deberá cumplir los siguientes requisitos:

- Utilizar Python y NumPy.
- Organizar el código mediante funciones con responsabilidades diferenciadas.
- Utilizar arrays de NumPy para almacenar y procesar los datos.
- Aplicar operaciones vectorizadas siempre que sea posible.
- Evitar repetir código.
- Utilizar nombres descriptivos para variables y funciones.
- Gestionar posibles errores durante la carga y el procesamiento de los datos.
- No utilizar Pandas, Scikit-learn ni bibliotecas externas de aprendizaje automático.

Se permite utilizar Matplotlib para representar gráficamente los resultados.

El programa deberá poder ejecutarse desde `main.py` sin necesidad de modificar manualmente el código.
