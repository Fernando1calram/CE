## 1. Programacion Orientada a Objetos

La Programacion Orientada a Objetos (POO) es un paradigma de programacion

### 1.1 Clases y objetos

instancia concreta de una clase.

**Conceptos clave:**
- `class` define una nueva clase
- Los **atributos** son variables que pertenecen a la clase
- Los **metodos** son funciones que pertenecen a la clase
- `self` es una referencia al objeto actual (como `this` en Java)

````python
class Perro:
    especie = "Canino"

    def __init__(self, nombre, raza, edad):
        self.nombre = nombre
        self.raza = raza
        self.edad = edad

    def ladrar(self):
        return f"{self.nombre} dice: Guau, guau"

    def descripcion(self):
        return f"{self.nombre}, {self.raza}, {self.edad} años"

rocky = Perro("Rocky", "Labrador", 3)
bella = Perro("Bella", "Pastor alemán", 5)
    
````

#### 1.1.2 Visibilidad de atributos

En Python, la visibilidad de los atributos funciona de forma diferente a Java o C++. Python no tiene public, private y protected como palabras clave.

En su lugar utiliza principalmente convenciones de nombres.

__PÚBLICO__

Por defecto todos los atributos son públicos y se puede acceder a ellos con la notacion punto `.`

```python
class Persona:
    def __init__(self, nombre):
        self.nombre = nombre

p = Persona("Juan")

print(p.nombre)
p.nombre = "Pedro"
```

__PROTEGIDO__

Para indicar que un atributo es protegido, por convención se antepone a su nombre un guión bajo `_variable`.

Es importante tener en cuenta que esta notación es de cortesía y nada impide acceder al atributo. Con la notación simplemente indicamos _"no me gustaría que accedieras a el"_

````python
class Persona:
    def __init__(self, nombre):
        self._nombre = nombre

p = Persona("Fernando")

print(p._nombre)
````

__PRIVADO__

Se indica que un atributo es protegido anteponiendo doble guión bajo `__variable`. Con esta nomenclatura, Python nos impide acceder al atributo, pero lo hace porque lo que realmente ocurre es que está renombrando el atributo a algo como `_Clase__atributo`

### 1.2 El metodo `__init__` y `self`

- **`__init__`**: es el **constructor** de la clase. Se ejecuta automaticamente
  cuando creas un objeto con `MiClase(...)`. No se llama explicitamente.
- **`self`**: es una referencia al **objeto actual**. Es necesario para acceder
  a los atributos y metodos del objeto desde dentro de la clase, por lo que siempre será el primer atributo que pongamos al definir un método. Se pasa automáticamente, no hace falta ponerlo al invocar el método.

> Nota: `self` no es una palabra reservada de Python, pero es la convencion
> universal. Si usas otro nombre, Python no lo reconocera.

````python
class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    def area(self):
        return self.ancho * self.alto

    def perimetro(self):
        return 2 * (self.ancho * self.alto)

r1 = Rectangulo(10,5)
r2 = Rectangulo(3,7)

print(f"Rectangulo 1: {r1.ancho}x{r1.alto}")
print(f" Area: {r1.area()}, Perimetro: {r1.perimetro()}")

print(f"Rectangulo 2: {r2.ancho}x{r2.alto}")
print(f" Area: {r2.area()}, Perimetro: {r2.perimetro()}")
````

````python
class Circulo:
    def __init__(self, radio=1.0):
        self.radio = radio

    def area(self):
        import math
        return math.pi * self.radio ** 2

    def perimetro(self):
        import math
        return 2 * math.pi * self.radio

c1 = Circulo()
print(f"Circulo 1 (radio=1): area= {c1.area():.4f}")

c2 = Circulo(5.0)
print(f"Circulo 2 (radio=5): area= {c2.area():.4f}")
````

### 1.3 Herencia simple y `super()`

La herencia permite crear una nueva clase a partir de una existente,
heredando todos sus atributos y metodos. La clase padre se llama
**superclase** y la nueva clase **subclase**.

Sintaxis Herencia simple: `class nombre_hijo(nombre_padre):`

**Beneficios:**
- Reutilizar codigo de clases existentes
- Extender funcionalidad sin modificar la clase original
- Crear jerarquias logicas de clases

En el caso de `super()` encontramos un par de diferencias significativas respecto a Java. Una la veremos más adelante, pero por el momento tenemos que tener en cuenta lo siguiente:
>En python, la llamada al método super() no tiene por qué ser la primera instrucción del método.

````python
class Animal:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def hablar(self):
        return "..."

    def descripcion(self):
        return f"{self.nombre}, {self.edad} años"

class Perro(Animal):
    def __init__(self, nombre, edad, raza):
        super().__init__(nombre, edad)
        self.raza = raza

    def hablar(self):
        return f"{self.nombre} dice guau!"

    def mostrar_raza(self):
        return f"Raza: {self.raza}"

class Gato(Animal):
    def __init__(self, nombre, edad, es_interior):
        super().__init__(nombre, edad)
        self.es_interior = es_interior

    def hablar(self):
        return f"{self.nombre} dice miau"

perro1 = Perro("Fernando", 5, "Prueba")
print(f"{perro1.hablar()}")

gato1 = Gato("FernandoPrueba", 5, "PruebaInterior")
print(f"{gato1.hablar()}")

roky = Perro("Rocky", 3, "Labrador")
bella = Gato("Bella",5,True)

print(f"{roky.nombre}: {roky.hablar}")
print(f" Raza: {roky.mostrar_raza()}")
print(f"Descripcion (heredada): {roky.descripcion()}")

print(f"{bella.nombre}: {bella.hablar}")
print(f" Es interior: {bella.es_interior}")
print(f"Descripcion (heredada): {bella.descripcion()}")
````

### 1.4 Metodos magicos basicos

Los métodos mágicos (o dunder methods) son metodos que empiezan y
terminan con doble guion bajo (`__`). Python los llama automáticamente
en situaciones específicas.

**Metodos magicos mas usados:**
- `__init__`: Constructor
- `__str__`: Representación en texto (para humanos)
- `__repr__`: Representación oficial (para depurar)
- `__len__`: Longitud de un objeto
- `__add__`: Suma (operador +)
- `__eq__`: Comparación de igualdad (==)

````python
class Libro:
    def __init__(self, titulo, autor, paginas):
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas

    def __str__(self):
        return f"Libro: {self.titulo} por {self.autor} tiene {self.paginas} páginas"

    def __repr__(self):
        return f"Libro ({self.titulo!r}, {self.autor!r}, {self.paginas})"

libro = Libro("1984", "George Orwell", 328)

print(libro)
print(repr(libro))
````

````python
class Playlist:
    def __init__(self, canciones=None):
        if canciones is None:
            self.canciones = []
        else:
            self.canciones = canciones

    def agregar(self, cancion):
        self.canciones.append(cancion)

    def __len__(self):
        return len(self.canciones)

    def __str__(self):
        return f"Playlist con {len(self)} canciones"

playlist = Playlist()
playlist.agregar("Bohemian Rhapsody")
playlist.agregar("Imagine")
playlist.agregar("Hotel California")
print(playlist)
print(f"Numero de canciones: {len(playlist)}")
````

````python
class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __add__(self, otro):
        return Vector(self.x + otro.x, self.y + otro.y)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(1,2)
v2 = Vector(3,4)
v3 = v1 + v2

print(f"{v1} + {v2} = {v3}")

v4 = v1 +v2 + v1
print(f"Suma encadenada: {v4}")
````

````python
class Punto:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __eq__(self, otro):
        if not isinstance(otro, Punto):
            return False
        return self.x == otro.x and self.y == otro.y

p1 = Punto(3,5)
p2 = Punto(5,8)
p3 = Punto(3,5)

print(p1.__eq__(p2))
print(p1.__eq__(p3))
````

### 1.5 Más Métodos mágicos

Os dejo aqui una lista de referencia de distintos métodos mágicos que ofrece python con su propósito. Aunque, como siempre, podéis acudir a la documentación oficial para más información.

__Constructor y Ciclo de Vida__

| Método | Cuándo se ejecuta | Ejemplo |
|--------|-------------------|---------|
| `__init__(self, ...)` | Al crear un objeto | `obj = Clase(1, 2)` |
| `__new__(cls, ...)` | Antes de `__init__`, crea la instancia | Interna |
| `__del__(self)` | Al eliminar el objeto (garbage collector) | `del obj` |

__Representación en Texto__

| Método | Cuándo se ejecuta | Ejemplo |
|--------|-------------------|---------|
| `__str__(self)` | `print(obj)`, `str(obj)` | `"Hola"` |
| `__repr__(self)` | `repr(obj)`, REPL, listas | `"Clase(1, 2)"` |
| `__bytes__(self)` | `bytes(obj)` | `b'...'` |

> **Diferencia clave:** `__str__` es para **humanos** (legible). `__repr__` es para **desarrolladores** (información completa, ideal para depurar). Si solo implementas uno, Python usa `__repr__` como fallback.

__Comparación__

| Método | Operador | Ejemplo |
|--------|----------|---------|
| `__eq__(self, otro)` | `==` | `a == b` |
| `__ne__(self, otro)` | `!=` | `a != b` |
| `__lt__(self, otro)` | `<` | `a < b` |
| `__le__(self, otro)` | `<=` | `a <= b` |
| `__gt__(self, otro)` | `>` | `a > b` |
| `__ge__(self, otro)` | `>=` | `a >= b` |
| `__hash__(self)` | `hash(obj)` (claves de diccionario) | `hash(obj)` |

__Operadores Aritméticos__

| Método | Operador | Ejemplo |
|--------|----------|---------|
| `__add__(self, otro)` | `+` | `a + b` |
| `__sub__(self, otro)` | `-` | `a - b` |
| `__mul__(self, otro)` | `*` | `a * b` |
| `__truediv__(self, otro)` | `/` | `a / b` |
| `__floordiv__(self, otro)` | `//` | `a // b` |
| `__mod__(self, otro)` | `%` | `a % b` |
| `__pow__(self, otro)` | `**` | `a ** b` |
| `__neg__(self)` | `-obj` | `-a` |
| `__abs__(self)` | `abs(obj)` | `abs(a)` |

__Operadores In-Place (modifican el objeto)__

| Método | Operador | Ejemplo |
|--------|----------|---------|
| `__iadd__(self, otro)` | `+=` | `a += b` |
| `__isub__(self, otro)` | `-=` | `a -= b` |
| `__imul__(self, otro)` | `*=` | `a *= b` |
| `__itruediv__(self, otro)` | `/=` | `a /= b` |

__Contenedores y Secuencias__



| Método | Propósito | Operación |
|--------|-----------|-----------|
| `__len__(self)` | Obtener la longitud del objeto | `len(obj)` |
| `__getitem__(self, key)` | Acceder a un elemento por clave/índice | `obj[key]` |
| `__setitem__(self, key, val)` | Asignar un valor por clave/índice | `obj[key] = val` |
| `__delitem__(self, key)` | Eliminar un elemento por clave/índice | `del obj[key]` |
| `__iter__(self)` | Hacer que el objeto sea iterable (para `for`) | `for x in obj:` |
| `__next__(self)` | Obtener el siguiente elemento en una iteración | `next(obj)` |
| `__contains__(self, item)` | Comprobar si un elemento está dentro del objeto | `"a" in obj` |

__Llamada y Conversión__

| Método | Propósito | Operación |
|--------|-----------|-----------|
| `__call__(self, ...)` | Hacer que un objeto se comporte como una función | `obj(args)` |
| `__int__(self)` | Convertir el objeto a un número entero | `int(obj)` |
| `__float__(self)` | Convertir el objeto a un número decimal | `float(obj)` |
| `__bool__(self)` | Determinar si el objeto se considera verdadero o falso | `bool(obj)` |
| `__format__(self, fmt)` | Personalizar cómo se formatea el objeto en un f-string | `f"{obj}"` |
| `__index__(self)` | Usar el objeto como un índice (entero) para secuencias | `obj[5:]` |

__Context Managers__

| Método | Cuándo se ejecuta | Ejemplo |
|--------|-------------------|---------|
| `__enter__(self)` | Inicio de `with` | `with obj:` |
| `__exit__(self, exc_type, exc_val, exc_tb)` | Fin de `with` | `with obj:` |

__Acceso a Atributos__

| Método | Cuándo se ejecuta | Ejemplo |
|--------|-------------------|---------|
| `__getattr__(self, name)` | `obj.atributo` (no encontrado) | `obj.x` |
| `__getattribute__(self, name)` | `obj.atributo` (siempre) | `obj.x` |
| `__setattr__(self, name, value)` | `obj.atributo = value` | `obj.x = 5` |
| `__delattr__(self, name)` | `del obj.atributo` | `del obj.x` |
| `__dir__(self)` | `dir(obj)` | `dir(obj)` |

#### 1.6.1 MRO (Method Resolution Order)

La primera gran duda que surge es que ocurrirá si nos encontramos con que varias de las clases de las que heredamos tienen uno o varios métodos con el mismo nombre.

En este caso entra en juego el __MRO__ _(Orden de Resolución de Métodos)_, que se comportará de la siguiente manera:

> Hijo → Superclase1 → Superclase2 → ... → SuperclaseN → object

Si al invocar un método lo encuentra en la propia clase, tomará ese. Si no, lo buscará en la primera superclase que aparece en la declaración. Si tampoco lo encuentra, continuará con la segunda superclase y así hasta que encuentre la primera ocurrencia.

#### 1.6.2 MRO y `super()`

Antes hablábamos de dos diferencias fundamentales de `super()` en python respecto a Java. Habíamos visto que `super()` no tenía por que aparecer como primera instrucción, sino que puede estar en cualquier posición. 

La segunda gran diferencia y la más importante por los errores que puede acarrear es:
> En python, `super()` no significa _llama a mi padre_, sino __llama a la siguiente clase en el MRO__

Y esto es de gran importancia en un escenario de herencia múltiple y puede dar lugar a errores al venir de otros lenguajes.

Viendo el siguiente código, si pensamos en términos de Java, la salida esperada sería
> D C A

Puesto que al invocarlo desde D, el método `__init__` se encuentra en C y dejaría de buscar. Y al llegar a la clase A no hereda de ninguna de las clases y, por tanto, su invocación a `super().__init__()` iría a parar a la clase `object`.

Pero si atendemos al MRO, tenemos que el orden sería:

> D (clase hija) -> C (padre) -> A (padre) -> B (madre)

Es decir, que cada vez que invocamos `super().__init__()` no estamos diciendo _"Llama a este método de mi padre"_ sino que estamos diciendo _"Llama a este método en la siguiente clase del MRO"_

### 1.8 Decoradores
Los decoradores en Python son una forma de modificar o ampliar el comportamiento de una función o clase sin tener que modificar directamente su código. 

Para entender su funcionamiento, tenemos que tener claro que en python una función también es un objeto. De tal modo que podemos hacer lo siguiente:

```python
def saludar():
    print("Hola")

f = saludar
f()
```
Pues ahora la variable `f`pasa a apuntar al mismo objeto que la función `saludar`.

Del mismo modo, podemos pasar una función como parámetro a otra función:

```python
def ejecutar(funcion):
    funcion()

ejecutar(saludar)
```

En este caso, podemos decir que ejecutar es un decorador, pues es una función que recibe otra función. El uso de la `@` en los decoradores que vamos a ver no es más que una manera más comoda de escribirlo.

Esto es más sencillo de entender si tenemos en cuenta que python está escrito en C y en C ya teníamos punteros a funciones.

#### 1.8.1 @property
Permite definir un método como si fuera un atributo de lectura. Es decir, accedes a él con la sintaxis de atributo (obj.valor), pero por debajo se ejecuta un método que puede hacer cálculos, validaciones, o lógica adicional.

__¿Para qué sirve?__
- __Encapsulación controlada__: controlar cómo se lee un atributo sin cambiar la interfaz pública.
- __Cálculos perezosos__ _(lazy evaluation)_: calcular un valor solo cuando se solicita.
- __Validación en lectura__: verificar condiciones antes de devolver un valor.
- __Compatibilidad con API existente__: cambiar un atributo por un cálculo sin romper código cliente.

La sintaxis para crear _getter_ y _setter_ (y _deleter_) para un atributo sería:

```python
@property
def nombre(self):
    ...

@nombre.setter
def nombre(self, valor):
    ...

@nombre.deleter
def nombre(self):
    ...
```

No obstante, _deleter_ se usa mucho menos que los _getter_ y _setter_ y quizá no siempre lo queráis implementar.

__REGLAS QUE DEBES TENER EN CUENTA__

| Regla | Detalle |
|-------|---------|
| **Nombre del setter** | Debe ser el mismo que el `@property` |
| **No usar el nombre como `self.nombre`** dentro del getter/setter | Causa recursión infinita. Usa el atributo con guion bajo: `self._nombre` |
| **Solo se accede como atributo** | No uses paréntesis al leer |
| **Úsalo para lógica ligera** | Si la operación es costosa, considera caché (`@functools.cached_property`) |

### 1.9 Clases Abstractas — Interfaces para modelos
Las **clases abstractas** permiten definir una interfaz que todas las subclases deben implementar. En el caso de Python, para hacer uso de clases abstractas debemos recurrir al módulo `abc` _(abstract base classes)_.

En concreto deberemos importar los siguientes elementos:

- clase `ABC`
- decoradores `abstractmethod` si vamos a incluir métodos abstractos y `abstractproperty` si vamos a incluir atributos abstractos

Ejemplo:

```python
from abc import ABC, abstractmethod

# Clase abstracta: hereda de ABC
class Forma(ABC):
    def __init__(self, color: str):
        self.color = color

    # Método concreto: todas las subclases lo heredan
    def describir_color(self) -> str:
        return f"Color: {self.color}"
        
    # Método abstracto: OBLIGATORIO implementar
    @abstractmethod
    def area(self) -> float:
        """Calcula el área de la forma."""
        pass

    # Otro método abstracto
    @abstractmethod
    def perimetro(self) -> float:
        """Calcula el perímetro de la forma."""
        pass

    # Método que usa los abstractos (template method)
    def info(self) -> str:
        return (f"{self.__class__.__name__} | "
                f"color={self.color} | "
                f"area={self.area():.2f} | "
                f"perímetro={self.perimetro():.2f}")
```

### 1.10 Iteradores

Un iterador es un objeto que permite recorrer una secuencia de elementos uno a uno, sin necesidad de tener todos los elementos disponibles simultáneamente.

La idea fundamental gira alrededor de dos métodos:
- __iter__() → devuelve un iterador.
- __next__() → devuelve el siguiente elemento.
- Cuando no quedan elementos, __next__() lanza __StopIteration__

En el tema anterior vimos los bucles en Python. Como ocurre en otros lenguajes como PHP o PL/SQL, el bucle `for` utiliza internamente un iterador. De tal modo que podríamos aproximar que:

```python
for numero in [10, 20, 30]:
    print(numero)
```

Realmente hace de manera interna algo similar a:

```python
iterador = iter([10, 20, 30])

while True:
    try:
        numero = next(iterador)
        print(numero)
    except StopIteration:
        break
```

#### 1.10.1 Iteradores personalizados — `__iter__` y `__next__`

El primer concepto que debe quedar claro es el de __iterador__ vs __iterable__. 
- __iterable__: es algo que se puede recorrer
- __iterador__: el objeto que mantiene el estado del recorrido.

Para que un objeto sea __iterable__, su clase debe tener definido el método `__iter__()`, que devolverá el identificador del objeto que tiene el iterador.

Ejemplo de `__iter__()`:

```python
   def __iter__(self):
        return self
```

Para que un objeto sea iterador, su clase debe implementar el método `__next__()`, que devolverá el siguiente elemento (o `StopIteration` si ha llegado al final):

Ejemplo de `__next__()`:
```python
    def __next__(self):
        if self.actual >= self.limite:
            raise StopIteration

        valor = self.actual
        self.actual += 1
        return valor
```
