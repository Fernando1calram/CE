### 1.4. Entornos Virtuales
Como habréis observado, la pestaña __Enviroments__ nos permite crear nuevos entornos. ¿En qué consiste esto? Imaginad que estamos trabajando en un proyecto que utiliza la última versión de python y varios paquetes actualizados a su última versión. Ahora nos pasan otro proyecto para que solucionemos algunos fallos o añadamos alguna funcionalidad, pero ese proyecto utiliza una versión anterior de python o de las bibliotecas. Si bajamos la versión de nuestro intérprete de python o de las bibliotecas, podremos trabajar en el proyecto recién llegado, pero ya no serán compatibles con el otro proyecto en el que estábamos. Para solucionar esto, python nos permite crear entornos virtuales. 

Mediante el uso de entornos virtuales podemos tener distintas versiones de python o de los paquetes para cada entorno y, para hacer uso de ellos, bastará con cambiar de un entorno a otro. Python nos provee del módulo `venv` para crear entornos virtuales y, además, existen otras utilidades más completas como `pipenv`o `poetry`. En la siguiente tabla podemos ver una comparativa entre ellos:

| Característica | **venv** (Estándar) | **pipenv** | **poetry** |
| :--- | :--- | :--- | :--- |
| **Gestión de Entornos** | Manual (`python -m venv`) | Automática (se crea al instalar) | Automática (se crea al instalar) |
| **Gestión de Dependencias** | Manual (`pip install`) | Automática (gestiona versiones) | Automática (gestiona versiones y resolución) |
| **Archivo de Configuración** | `requirements.txt` (lista plana) | `Pipfile` + `Pipfile.lock` | `pyproject.toml` + `poetry.lock` |
| **Bloqueo de Versiones (Lock)** | ❌ No (manual) | ✅ Sí (`Pipfile.lock`) | ✅ Sí (`poetry.lock`) |
| **Resolución de Dependencias** | ❌ No (puede haber conflictos) | ✅ Sí (búsqueda inteligente) | ✅ Sí (algoritmo más robusto) |
| **Gestión de Python** | ❌ No (requiere Python instalado) | ❌ No (requiere Python instalado) | ❌ No nativamente (pero se integra con pyenv) |
| **Reproducibilidad** | ❌ Baja (versiones flotantes) | ✅ Alta | ✅ Muy Alta |
| **Velocidad** | Muy rápida | Lenta (resolución lenta) | Moderada  |
| **Curva de Aprendizaje** | Baja (básico) | Media (concepto de Pipfile) | Media (pyproject.toml) |
| **Adopción en la Industria** | Alta (básico) | Media (estancada) |  En aumento (estándar moderno) |
| **Comunidad/Ecosistema** | Nativo de Python | Activa pero ralentizada | Muy activa (crece rápido) |
| **Ideal para** | Scripts simples, aprendizaje | Proyectos medianos | **Proyectos profesionales/producción** |

No obstante, por simplicidad, nosotros vamos a utilizar la gestión que hace Anaconda de los entornos.

Crear y eliminar entornos virtuales con Anaconda, así como instalar o eliminar paquetes es algo que podemos hacer cómodamente a través de la interfaz. Y dándole al botón de `play` podremos lanzar un terminal con el entorno activado.

### 1.5 Jupyter Notebooks
Jupyter Notebooks es una herramienta que nos permite crear cuadernos de trabajo en formato Markdown con código python embebido. Esto resulta particularmente interesante en entornos educativos y de investigación.

Esta herramienta nos permite, además de incluir bloques de texto en formato Markdown, incluir bloques de texto que se pueden ejecutar aisladamente y en el orden que queramos. Por supuesto, en caso de que haya dependencias de código entre los bloques, estos darán un error.

Hagamos una prueba. Para ejecutar cada bloque, seleccionalo y pulsa `Crtl+Enter`

Ejecuta ahora el siguiente bloque _(Instruccion 1)_:

````````python
#Instruccion 1
print(dato)
````````

Como ves, la instrucción da un error porque no existe la variable dato, no ha sido creada aún. Ejecutemos el siguiente bloque _(Instruccion 2)_:

````````python
#Instruccion 2
dato = 'Hola Mundo'
````````

Ahora vuelve a ejecutar el bloque de la _Instruccion 1_, el que contenía la instrucción `print`.

Ya no muestra ningún error y obtenemos la salida de la ejecución, porque hemos creado antes la variable `dato`. Puede parecer confuso dado que la variable se está creando, aparentemente, una linea después. Aunque esto realmente no es así. Junto a la linea de código, puedes ver unos corchetes con un número `[x]:`, ese número indica el orden de ejecución. Si el número de la línea _Instruccion 2_ es inferior al de la línea _Instruccion 1_ es que se ha ejecutado antes.

Obviamente este es un ejemplo forzado para entender el funcionamiento y, para evitar severos dolores de cabeza, seguiremos un orden en el código. Sin embargo, ejecutar bloques de manera aislada nos permite que, ante errores o cambios, no tengamos que volver a ejecutar todo el código y, si hemos realizado algún cálculo complejo o que consume recursos, no tengamos que volver a ejecutarlo si estamos modificando código posterior.

Podemos lanzar __jupyter notebooks__ tanto desde __Anaconda__ como desde una __terminal__ escribiendo: `jupyter notebook`

## 3. Sintaxis Básica de Python
Python es un lenguaje de programación creado por Guido van Rossum y lanzado en 1991. El objetivo era crear un lenguaje de script ligero y con una sintaxis sencilla y fácil de aprender. Para ello, lo primero que desaparece de Python son los delimitadores.

- Los `;` al final de cada sentencia no son necesarios en Python _(se recomienda no hacer uso de ellos)_
- Las instrucciones compuestas, como clases, métodos, funciones o estructuras de control no hacen uso de `{` y `}`. En su lugar el peso recae sobre una férrea identación. La necesidad de usar tabulación, también contribuye a la legibilidad.

Además, las aplicaciones Python __carecen de método `main`__

Veamos un ejemplo sencillo de código:

````````python
# funcion para sumar dos elementos
def suma(a,b):
    return a+b

resultado = suma(4, 5)

if resultado < 10:
    print(f'{resultado} es menor que 10')
else: 
    print('Numero mayor que 10')
````````

### 3.1. Variables
Una **variable** es un espacio en memoria que almacena un valor. En Python, no es necesario declarar el tipo de la variable; se infiere automáticamente según el valor que se le asigna.

#### Sintaxis

```python
nombre_variable = valor
```

#### Reglas para nombrar variables
- Deben empezar con una letra o guion bajo (`_`)
- No pueden empezar por un número
- Solo contienen letras, números y guiones bajos
- Son **case-sensitive** (`Edad` ≠ `edad`)
- No pueden ser palabras reservadas (`if`, `for`, `def`, etc.)
- Convención recomendada: `nombre_con_guion_bajo` (snake_case)

````````python
# Ejemplos de variables
nombre = "Ana"           # string (texto)
edad = 25                # integer (número entero)
altura = 1.68            # float (número decimal)
es_estudiante = True     # boolean (verdadero/falso)

print(nombre)
print(edad)
print(altura)
print(es_estudiante)
````````

````````python
# Asignación múltiple
x, y, z = 10, 20, 30
print(f"x={x}, y={y}, z={z}")

# Asignación a una misma variable
a = b = c = 100
print(f"a={a}, b={b}, c={c}")
````````

````````python
# Función type() para conocer el tipo de una variable
nombre = "Ana"
edad = 25
altura = 1.68
es_estudiante = True

print(type(nombre))   # <class 'str'>
print(type(edad))     # <class 'int'>
print(type(altura))   # <class 'float'>
print(type(es_estudiante))  # <class 'bool'>
````````

### 3.2. Tipos de datos básicos

Python tiene varios tipos de datos integrados. Los más importantes son:

| Tipo | Nombre en Python | Ejemplo | Descripción |
|------|-----------------|---------|-------------|
| Entero | `int` | `42`, `-7` | Números sin decimales |
| Decimal | `float` | `3.14`, `-0.5` | Números con decimales |
| Texto | `str` | `"Hola"`, `'Python'` | Secuencia de caracteres |
| Booleano | `bool` | `True`, `False` | Valores verdadero/falso |
| Nulo | `NoneType` | `None` | Ausencia de valor |

````````python
# Enteros (int)
entero_positivo = 10
entero_negativo = -3
print(f"Enteros: {entero_positivo}, {entero_negativo}")

# Decimales (float)
precio = 19.99
pi = 3.14159
print(f"Decimales: {precio}, {pi}")

# Strings (str)
saludo = "Hola Mundo"
curso = 'Programación de IA'
print(f"Strings: {saludo}, {curso}")

# Booleanos (bool)
aprobado = True
suspenso = False
print(f"Booleanos: {aprobado}, {suspenso}")
````````

### 3.3. Operadores

Los operadores permiten realizar operaciones sobre variables y valores. En Python se clasifican en:

#### 3.3.1. Operadores aritméticos

| Operador | Descripción | Ejemplo | Resultado |
|----------|-------------|---------|-----------|
| `+` | Suma | `5 + 3` | `8` |
| `-` | Resta | `10 - 4` | `6` |
| `*` | Multiplicación | `4 * 2` | `8` |
| `/` | División (decimal) | `10 / 3` | `3.333...` |
| `//` | División entera | `10 // 3` | `3` |
| `%` | Módulo (residuo) | `10 % 3` | `1` |
| `**` | Potencia | `2 ** 3` | `8` |

````````python
# Operadores aritméticos
a = 15
b = 4

print(f"Suma:         {a} + {b} = {a + b}")
print(f"Resta:        {a} - {b} = {a - b}")
print(f"Multiplicación: {a} * {b} = {a * b}")
print(f"División:     {a} / {b} = {a / b}")
print(f"División ent.: {a} // {b} = {a // b}")
print(f"Módulo:       {a} % {b} = {a % b}")
print(f"Potencia:     {a} ** {b} = {a ** b}")
````````

#### 3.3.2. Operadores de comparación

Comparan dos valores y devuelven un booleano (`True` o `False`).

| Operador | Descripción | Ejemplo | Resultado |
|----------|-------------|---------|-----------|
| `==` | Igual | `5 == 5` | `True` |
| `!=` | Diferente | `5 != 3` | `True` |
| `>` | Mayor que | `5 > 3` | `True` |
| `<` | Menor que | `3 < 5` | `True` |
| `>=` | Mayor o igual | `5 >= 5` | `True` |
| `<=` | Menor o igual | `3 <= 5` | `True` |

````````python
# Operadores de comparación
x = 10
y = 20

print(f"{x} == {y}: {x == y}")    # False
print(f"{x} != {y}: {x != y}")    # True
print(f"{x} >  {y}: {x > y}")     # False
print(f"{x} <  {y}: {x < y}")     # True
print(f"{x} >= {x}: {x >= x}")    # True
print(f"{y} <= {y}: {y <= y}")    # True
````````

#### 3.3.3. Operadores lógicos

Se usan para combinar expresiones booleanas.

| Operador | Descripción | Ejemplo |
|----------|-------------|---------|
| `and` | True si AMBOS son True | `True and False` → `False` |
| `or`  | True si ALGÚN es True  | `True or False` → `True` |
| `not` | Invierte el valor       | `not True` → `False` |

````````python
# Operadores lógicos
edad = 25
tiene_titulo = True
tiene_experiencia = False

# and: ambas condiciones deben ser True
puede_trabajar = (edad >= 18) and (tiene_titulo)
print(f"Puede trabajar (edad >= 18 Y título): {puede_trabajar}")

# or: basta una condición True
puede_entrar = (edad >= 18) or (tiene_experiencia)
print(f"Puede entrar (edad >= 18 O experiencia): {puede_entrar}")

# not: invierte el valor
no_experiencia = not tiene_experiencia
print(f"No tiene experiencia: {no_experiencia}")
````````

### 3.4. Entrada y salida (I/O)

#### 3.4.1. `print()` — Salida de datos

La función `print()` muestra información en pantalla. Se puede usar para mostrar textos, variables o el resultado de expresiones.

#### Formatos de impresión

- **Formato básico**: `print(valor)`
- **Concatenación**: `print("Texto " + str(variable))`
- **f-strings** (recomendado): `print(f"Texto {variable}")`

````````python
# print() — Formas de imprimir

# 1. Imprimir texto literal
print("Hola Mundo")

# 2. Imprimir variables
nombre = "Carlos"
edad = 22
print(nombre)
print(edad)

# 3. Imprimir múltiples valores (separados por coma)
print("Nombre:", nombre, "Edad:", edad)

# 4. f-strings (formato recomendado)
print(f"{nombre} tiene {edad} años")

# 5. Concatenación con +
print("Nombre: " + nombre + ", Edad: " + str(edad))
````````

#### 3.4.2. `input()` — Entrada de datos

La función `input()` permite al usuario escribir datos desde el teclado. **Siempre devuelve un string**, por lo que es necesario convertirlo al tipo deseado.

````````python
# input() — Capturar datos del usuario

# 1. Capturar un texto
nombre_usuario = input("¿Cómo te llamas? ")
print(f"Encantado de conocerte, {nombre_usuario}")

# 2. Capturar un número (recordar convertir!)
# edad = int(input("¿Cuántos años tienes? "))
# print(f"Tienes {edad} años")
````````

### 3.5. Conversión de tipos (Casting)

A veces necesitamos cambiar el tipo de una variable. Python proporciona funciones para convertir entre tipos:

| Función | Tipo destino | Ejemplo |
|---------|-------------|---------|
| `int()` | Entero | `int("42")` → `42` |
| `float()` | Decimal | `float("3.14")` → `3.14` |
| `str()` | Texto | `str(42)` → `"42"` |
| `bool()` | Booleano | `bool(1)` → `True` |
| `type()` | — | `type(42)` → `<class 'int'>` |

````````python
# Conversión de tipos

# int() — Convertir a entero
print(int(3.9))       # 3 (trunca, no redondea)
print(int("42"))      # 42

# float() — Convertir a decimal
print(float(5))       # 5.0
print(float("3.14"))  # 3.14

# str() — Convertir a texto
print(str(100))       # "100"
print(str(3.14))      # "3.14"

# bool() — Convertir a booleano
print(bool(1))        # True
print(bool(0))        # False
print(bool("Hola"))   # True
print(bool(""))       # False (cadena vacía)
print(bool([]))       # False (lista vacía)
````````

````````python
# Ejemplo práctico: sumar dos números introducidos por el usuario

# El usuario introduce dos números como texto
num1_str = input("Introduce el primer número: ")
num2_str = input("Introduce el segundo número: ")

# Convertimos a float (para permitir decimales)
num1 = float(num1_str)
num2 = float(num2_str)

# Calculamos la suma
resultado = num1 + num2

# Mostramos el resultado
print(f"El resultado de {num1} + {num2} es {resultado}")
````````
