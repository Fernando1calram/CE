#### 1.1.3 Operador ternario

Permite escribir un if/else en una sola linea.

Sintaxis: 
> `valor_si_verdadero`__`if`__`condicion`__`else`__`valor_si_falso`

````python
# Comparacion: forma larga vs. ternaria
edad = 16

# Forma larga
if edad >= 18:
    acceso = "permitido"
else:
    acceso = "denegado"
print(f"Forma larga: {acceso}")

# Forma ternaria (equivalente)
acceso = "permitido" if edad >= 18 else "denegado"
print(f"Forma corta: {acceso}")

# Ejemplo: valor absoluto sin abs()
numero = -42
absoluto = numero if numero >= 0 else -numero
print(f"Valor absoluto de {numero}: {absoluto}")
````

### 1.2 Bucles: for y while

El bucle __for__ itera **sobre elementos** _(no sobre contadores como en C)_.

Sintaxis 
>`for`_`variable`_` in `_`coleccion`_`:`

El bucle while repite mientras una condicion sea verdadera.

Sintaxis 
>`while`_`expresion_logica`_`:`

````python
# for con listas
frutas = ["manzana", "platano", "pera", "naranja"]
for fruta in frutas:
    print(f"Me gusta la {fruta}")

# for con strings (caracter por caracter)
print()
for letra in "Python":
    print(letra, end=" ")
print()

# for con range()
print("range(5):", end=" ")
for i in range(5):
    print(i, end=" ")
print()

print("range(2, 8):", end=" ")
for i in range(2, 8):
    print(i, end=" ")
print()

print("range(0, 10, 2):", end=" ")
for i in range(0, 10, 2):
    print(i, end=" ")
print()
````

````python
# for con diccionarios
edades = {"Ana": 25, "Carlos": 30, "Beatriz": 22}

# Iterar sobre claves
for nombre in edades:
    print(f"  {nombre}: {edades[nombre]}")

# Iterar sobre pares (recomendado)
print()
for nombre, edad in edades.items():
    print(f"  {nombre} tiene {edad} años")

# while loop
print("Contandor con while:")
contador = 0
while contador < 5:
    print(f"  {contador}", end=" ")
    contador += 1
print()
````

#### 1.2.1 La función range() en detalle

| Sintaxis | Que hace |
|----------|----------|
| range(stop) | 0, 1, ..., stop-1 |
| range(start, stop) | start, start+1, ..., stop-1 |
| range(start, stop, step) | start, start+step, ..., hasta stop

````python
# Invertir un string con range
texto = "Python"
invertido = ""
for i in range(len(texto) - 1, -1, -1):
    invertido += texto[i]
print(f"'{texto}' invertido -> '{invertido}'")

# Suma de los primeros 100 números
suma = 0
for i in range(1, 101):
    suma += i
print(f"Suma de 1 a 100: {suma}")  # 5050

# Iterar con índice usando enumerate()
frutas = ["manzana", "platano", "pera"]
for i, fruta in enumerate(frutas):
    print(f"  [{i}] {fruta}")
````

#### 1.2.2 Control de bucles: break, continue, pass

| Sentencia | Accion |
|-----------|--------|
| break | Sale completamente del bucle |
| continue | Salta al siguiente ciclo |
| pass | No hace nada (marcador de posicion)

````python
# break: salir del bucle al encontrar un valor
numeros = [3, 7, 1, 9, 4, 8]
for n in numeros:
    if n == 9:
        print(f"Encontrado: {n}!")
        break
    print(n, end=" ")
print()

# continue: saltar al siguiente ciclo
print("Numeros del 1 al 10, solo pares:")
for i in range(1, 11):
    if i % 2 != 0:
        continue
    print(i, end=" ")
print()

# pass: marcador de posicion (se reemplazará con lógica)
print("Ejemplo de pass:")
for letra in "Hola":
    if letra == "l":
        pass  # <- se reemplazará con lógica
    else:
        print(letra, end="")
print()
````

## 2. Mutable e Inmutable

Llegados es este punto es importante comentar que los tipos de datos en python se clasifican en dos tipos: __Mutables__ e __Inmutables__

- __Inmutables__: No se pueden modificar. Para cambiar el valor de la variable hay que crear una nuev. P.ej: Tipos básicos, tuplas.
__Mutables__: Se pueden modificar. P.ej: Listas, Diccionarios, Conjuntos.

Vamos a verlo con un ejemplo:

```python
a = 10
b = a
```
```
a ──┐
    ├──► 10
b ──┘
```

Ambas variables están apuntando al mismo dato.

Vamos a modificar `b`

```python
b = 20
```
```
a ─────► 10

b ─────► 20
```

Ahora cada variable apunta a un dato distino. Un int es inmutable, por lo tanto no se ha modificado el valor 10, sino que se ha creado un nuevo objeto.

>La asignación en Python no copia el objeto: hace que otro nombre lo referencie. Si el objeto es mutable, ambos nombres pueden observar las modificaciones del mismo objeto. Si es inmutable, no podemos modificar ese objeto, sino que reasignamos el nombre a otro objeto.

Veamos el mismo ejemplo con un objeto mutable:
```python
a = [1, 2, 3]
b = a
```
```
       ┌───────────────┐
a ────►│ [1, 2, 3]     │
       └───────────────┘
             ▲
b ───────────┘
```
Ambas variables apuntan al mismo objeto. Ahora vamos a modificar `b`añadiendo un nuevo elemento.

```python
b.append(4)
```

El resultado que obtenemos es: 
```
       ┌──────────────────┐
a ────►│ [1, 2, 3, 4]     │
       └──────────────────┘
             ▲
b ───────────┘
```
Como el objeto es mutable, se ha modificado sin crear otro.

### ¿Por qué es esto importante? 

    1. Porque dos variables pueden modificar el mismo objeto
    2. Porque afecta a las funciones (paso de parámetros)
    3. Porque las copias no siempre son copias
    4. En IA trabajaremos constantemente con objetos mutables: 
        - listas
        - diccionarios
        - conjuntos
        - arrays
        - tensores
        - datasets
        - ...

## 3. Estructuras de Datos

Python ofrece cuatro estructuras de datos integradas fundamentales.

| Tipo | Delimitadores | Sintaxis | Ejemplo |

| **Lista** | `[ ]` | `[elemento1, elemento2, ...]` | `[1, 2, 3]` |

| **Conjunto** | `{ }` | `{elemento1, elemento2, ...}` | `{1, 2, 3}` |

### 3.1 Listas

Colecciones **ordenadas** y **mutables**. Son heterogéneas, es decir, pueden contener varios tipos de datos.

**Características clave:**
- Indexacion con [0], [-1] (último), slicing [inicio:fin:paso]
- Metodos: append(), insert(), pop(), remove(), sort(), len()

#### 3.1.1 Creación y acceso

````python
# Creación y acceso
numeros = [10, 20, 30, 40, 50]
mezcla = [1, "hola", 3.14, True]

print(f"Primero: {numeros[0]}")       # 10
print(f"Último: {numeros[-1]}")       # 50
print(f"Segundo último: {numeros[-2]}")  # 40
````

#### 3.1.2 Slicing
El slicing permite extraer subconjuntos de una lista (o cualquier secuencia) sin modificar la original.

Sintaxis: `lista[inicio : fin : paso]`

| Parámetro | Qué hace | Si se omite... |
|-----------|----------|----------------|
| `inicio`  | Índice donde **empieza** (incluido) | `0` (inicio de la lista) |
| `fin`     | Índice donde **termina** (excluido) | `len(lista)` (final de la lista) |
| `paso`    | Saltos entre elementos | `1` (uno a uno) |


````python
# Slicing: lista[inicio:fin:paso]
numeros = [10, 20, 30, 40, 50]
print(f"numeros[1:4] = {numeros[1:4]}")     # [20, 30, 40]
print(f"numeros[:3]  = {numeros[:3]}")      # [10, 20, 30]
print(f"numeros[::2] = {numeros[::2]}")     # [10, 30, 50]
print(f"numeros[::-1]= {numeros[::-1]}")    # invertida
````

#### 3.1.3 Metodos principales

| Método | Descripción | Ejemplo |
|--------|-------------|---------|
| `append()` | Añade un elemento al final | `lista.append(4)` |
| `insert()` | Añade en una posición específica | `lista.insert(1, 'a')` |
| `extend()` | Añade varios elementos (iterable) | `lista.extend([1, 2])` |
| `pop()` | Elimina y devuelve el último (o índice) | `lista.pop()` |
| `remove()` | Elimina la primera aparición de un valor | `lista.remove(2)` |
| `index()` | Devuelve el índice de un valor | `lista.index(3)` |

| `sort()` | Ordena la lista (in-place) | `lista.sort()` |
| `reverse()` | Invierte la lista (in-place) | `lista.reverse()` |
| `clear()` | Vacía la lista | `lista.clear()` |


````python
# Metodos principales
frutas = ["manzana", "pera"]
frutas.append("uva")           # al final
print(frutas)
frutas.insert(1, "naranja")    # en posicion
print(frutas)
frutas.remove("pera")          # por valor
print(frutas)
ultimo = frutas.pop()          # elimina y devuelve
print(frutas)
frutas.extend(["kiwi", "melon"])  # varios elementos
print(frutas)

print(f"Resultado: {frutas}")
print(f"Longitud: {len(frutas)}")
print(f"manzana in frutas: {"manzana" in frutas}")
````

#### 3.1.5 Comprensión de listas
La comprensión en python no es más que una sintaxis reducida para crear colecciones a partir de otras colecciones o iterables.

La sintaxis es: `[expresion`__`for`__` elemento `__`in`__`iterable]`

````python
# Sin comprensión
cuadrados = []

for numero in range(1, 6):
    cuadrados.append(numero ** 2)

print(f'Sin comprension: {cuadrados}')

# Con comprensión
cuadrados2 = [numero ** 2 for numero in range(1, 6)]
print(f'Con comprension: {cuadrados2}')
````

#### 3.1.6 Comprensión de listas con filtrado (condición)
Además podemos añadir una pequeña condición para hacer un filtrado de resultados.

La sintaxis es: `[expresion`__`for`__` elemento `__`in`__`iterable`__`if`__`condicion]`

````python
# Comprensión con condición
cuadrados3 = [numero ** 2 for numero in range(1, 6) if numero % 2 == 0]
print(f'Con comprension y condicion: {cuadrados3}')
````

#### 3.1.7 Comprensión de listas con expresiones ternarias
Puedes usar una expresion ternaria para transformar condicionalmente los elementos.

````python
# Clasificar numeros como par o impar
numeros = list(range(1, 11))
etiquetas = ["par" if n % 2 == 0 else "impar" for n in numeros]
print(f"Etiquetas: {list(zip(numeros, etiquetas))}")

# Convertir valores negativos a 0
valores = [-3, 5, -1, 8, -7, 2]
positivos = [v if v > 0 else 0 for v in valores]
print(f"Valores positivos: {positivos}")
````

#### 3.1.8 Matrices

````python
# Listas de listas (matrices simples)
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(f"Elemento [1][1]: {matriz[1][1]}")  # 5
print(f"Primera fila: {matriz[0]}")  # [1, 2, 3]
````

#### 3.1.9 Comprensión de listas anidada
También se pueden anidar comprensiones de listas para trabajar con matrices _(algo muy común en IA)_

````python
# Matriz (lista de listas)
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Aplanar una matriz (una sola lista)
aplanada = [num for fila in matriz for num in fila]
print(f"Aplanada: {aplanada}")

# Solo los numeros pares de la matriz
pares_matriz = [num for fila in matriz for num in fila if num % 2 == 0]
print(f"Pares en matriz: {pares_matriz}")

# Transponer una matriz
transpuesta = [[matriz[fila][col] for fila in range(len(matriz))] for col in range(len(matriz[0]))]
print(f"Transpuesta: {transpuesta}")

# Sumar los elementos de cada fila
sumas_fila = [sum(fila) for fila in matriz]
print(f"Sumas por fila: {sumas_fila}")
````

### 3.3 Diccionarios

Colecciones de **pares clave-valor**. Las claves deben ser inmutables y únicas.

Sintaxis: 
> `{clave1:valor1, clave2:valor2, ..., claveN:valorN}`

**Características:**
- Acceso por clave: diccionario[clave]
- .get(clave, valor_por_defecto) para acceso seguro
- .keys(), .values(), .items() para iterar

**Métodos**
| Método | Descripción | Ejemplo |
|--------|-------------|---------|
| `keys()` | Devuelve las claves | `d.keys()` |
| `values()` | Devuelve los valores | `d.values()` |
| `items()` | Devuelve pares (clave, valor) | `d.items()` |
| `get(key, default)` | Acceso seguro (no da error si falta) | `d.get('edad', 0)` |
| `pop(key, default)` | Elimina y devuelve el valor | `d.pop('nombre')` |
| `popitem()` | Elimina y devuelve el último par (LIFO) | `d.popitem()` |
| `update()` | Actualiza o añade pares | `d.update({'edad': 25})` |
| `setdefault(key, default)` | Añade si no existe, sino devuelve | `d.setdefault('ciudad', 'Madrid')` |
| `clear()` | Vacía el diccionario | `d.clear()` |
| `copy()` | Copia superficial del diccionario | `d2 = d.copy()` |
| `fromkeys()` | Crea un nuevo dict con claves dadas | `dict.fromkeys(['a', 'b'], 0)` |

#### 3.3.5 Comprensión de diccionarios
Las **dict comprehensions** crean diccionarios de forma compacta.

>Sintaxis: `{clave`__`:`__`valor`__`for`__`elemento`__`in`__`iterable}`

También pueden añadir condiciones:

>Sintaxis: `{clave`__`:`__`valor`__`for`__`elemento`__`in`__`iterable`__`if`__`expresion}`


````python
# Crear un diccionario con cuadrados
numeros = [1, 2, 3, 4, 5]
cuadrados = {n: n**2 for n in numeros}
print(f"Cuadrados: {cuadrados}")

# Invertir un diccionario (clave <-> valor)
inventario = {"manzana": 10, "platan": 5, "uva": 8}
invertido = {v: k for k, v in inventario.items()}
print(f"Invertido: {invertido}")

# Filtrar con condicion
mayores_5 = {k: v for k, v in inventario.items() if v > 5}
print(f"Con mas de 5 unidades: {mayores_5}")

# Transformar valores
precios = {"laptop": 999.99, "telefono": 699.99, "auriculares": 49.99}
descuento = {k: v * 0.9 for k, v in precios.items()}
print(f"Con 10% descuento: {descuento}")
````

#### 3.3.6 Zip (Empaquetar)
Combina varias listas (u otros iterables) elemento a elemento, creando un iterable de tuplas.

>Sintaxis: `zip(iterable1, iterable2, ...)`

| Operación | Resultado |
|-----------|-----------|
| `zip([1,2], [3,4])` | `[(1, 3), (2, 4)]` |
| `dict(zip(claves, valores))` | Diccionario combinado |
| `zip(a, b)` con longitudes distintas | Se trunca al iterable más corto |


````python
# Diccionario desde dos listas (zip)
nombres = ["manzana", "platano", "pera"]
precios = [1.2, 0.8, 2.5]
frutas_precios = dict(zip(nombres, precios))
print(f"Frutas y precios: {frutas_precios}")
````

__IMPORTANTE__: Dado que zip devuelve un iterable, debemos tener en cuenta que este __se consumirá al leerlo__. Si queremos evitar esto, debemos almacenarlo al recorrerlo, como estamos haciendo arriba al convertirlo en un diccionario.

Veamos el comportamiento al recorrerlo dos veces:

````python
# Sin almacenarlo
nombres = ["manzana", "platano", "pera"]
precios = [1.2, 0.8, 2.5]
frutas_precios = zip(nombres, precios)
print(f"Frutas y precios: {list(frutas_precios)}")
print(f"Volvemos a mostrarlo: {list(frutas_precios)}")
````

Y ahora veamos el comportamiento al almacenarlo al recorrerlo:

### 3.4 Conjuntos (set)

Colecciones **no ordenadas** y __mutables__ de elementos **únicos**. Ideales para
eliminar duplicados y operaciones matematicas.

Al igual que los diccionarios, los conjuntos de delimitan con `{}`. 

Sintaxis: 
> `{elemento1, elemento2, ..., elementoN}`

__IMPORTANTE__
> Un `{}` vacío en Python siempre es un diccionario. Para un conjunto vacío se usa `set()`

#### 3.4.1 Creacion

````python
# Creación
numeros_set = {1, 2, 3, 4, 5}
# Los duplicados se eliminan automáticamente
con_dupes = {1, 2, 2, 3, 3, 4, 5}
print(f"Set original: {numeros_set}")
print(f"Con duplicados: {con_dupes}")  # {1, 2, 3, 4, 5}

# Eliminar duplicados de una lista
lista_con_dupes = [1, 2, 2, 3, 3, 4, 5]
sin_dupes = set(lista_con_dupes)
print(f"Lista -> Set: {lista_con_dupes} -> {sin_dupes}")
````

#### 3.4.2 Operaciones de conjuntos

````python
# Operaciones de conjuntos
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print(f"A = {a}")
print(f"B = {b}")
print(f"Unión (A | B): {a | b}")           # todos los elementos
print(f"Intersección (A & B): {a & b}")     # solo comunes
print(f"Diferencia (A - B): {a - b}")       # solo en A
print(f"Diff. simétrica (A ^ B): {a ^ b}")  # en uno pero no ambos

# Metodos
mi_set = {1, 2, 3}
mi_set.add(4)
print(f"Después de add(4): {mi_set}")

mi_set.discard(99)  # sin error si no existe
print(f"Después de discard(99): {mi_set}")
````

#### 3.4.3 Comprensión de conjuntos

Crean conjuntos de forma compacta. Útiles para eliminar duplicados y operaciones matemáticas.

>Sintaxis: `{valor`__`for`__`elemento`__`in`__`iterable}`

También pueden añadir condiciones:

>Sintaxis: `{valor`__`for`__`elemento`__`in`__`iterable`__`if`__`expresion}`

````python
# Eliminar duplicados de una lista
numeros = [1, 2, 2, 3, 3, 4, 4, 5]
sin_dupes = {n for n in numeros}
print(f"Sin duplicados: {sin_dupes}")

# Caracteres unicos de un string
palabra = "banana"
unicos = {c for c in palabra}
print(f"Caracteres unicos: {unicos}")

# Numeros al cuadrado (conjunto)
cuadrados_set = {n**2 for n in range(1, 11)}
print(f"Cuadrados (set): {cuadrados_set}")

# Filtrar con condicion
pares_set = {n for n in range(1, 21) if n % 2 == 0}
print(f"Pares (set): {pares_set}")
````

## 4. Funciones

Las funciones permiten organizar código en bloques reutilizables. Son
fundamentales para escribir programas grandes y mantenibles.

Una de las peculiaridades de python es que las funciones pueden devolver más de un valor de retorno. Aunque esto es una verdad a medias, puesto que lo que devuelve realmente es una única tupla que se desempaqueta al recoger los valores.

### 4.1 Definicion basica: def, parametros, return

````python
# Función simple
def saludar(nombre):
    print(f"Hola, {nombre}!")

saludar("Ana")

# Función con valor de retorno
def calcular_area_circulo(radio):
    pi = 3.14159265
    return pi * radio ** 2

area = calcular_area_circulo(5)
print(f"
Área del círculo (r=5): {area:.2f}")

# Múltiples valores de retorno (devuelve una tupla)
def min_max(lista):
    return min(lista), max(lista)

menor, mayor = min_max([3, 7, 1, 9, 4])
print(f"Min: {menor}, Max: {mayor}")
````

### 4.2 Tipos de parametros

| Tipo | Sintaxis | Descripcion |
|------|----------|-------------|
| Posicional | def f(a, b) | Obligatorio, por posicion |
| Con valor por defecto | def f(a, b=10) | Opcional, usa valor si no se pasa |
| Por nombre | f(a=1, b=2) | Se especifica el nombre |
| *args | def f(*args) | Argumentos posicionales variables (tupla) |
| **kwargs | def f(**kwargs) | Argumentos con nombre variables (diccionario)

````python
# Parametros con valor por defecto
def presentar(nombre, edad=18, ciudad="Desconocida"):
    print(f"{nombre}, {edad} años, de {ciudad}")

presentar("Ana")                         # usa valores por defecto
presentar("Carlos", 25)                  # sobreescribe edad
presentar("Beatriz", 30, "Madrid")       # todos explícitos
presentar("Dani", ciudad="Barcelona")    # por nombre
````

````python
# *args: número variable de argumentos
def sumar_todos(*args):
    print(f"  args tipo: {type(args)}  -> {args}")
    return sum(args)

print(f"sumar_todos(1, 2): {sumar_todos(1, 2)}")
print(f"sumar_todos(1, 2, 3, 4): {sumar_todos(1, 2, 3, 4)}")
print(f"sumar_todos(): {sumar_todos()}")

# **kwargs: número variable de argumentos con nombre
def crear_perfil(**kwargs):
    print(f"  kwargs tipo: {type(kwargs)}  -> {kwargs}")
    return kwargs

perfil = crear_perfil(nombre="Ana", edad=22, ciudad="Madrid", carrera="IA")
print(f"Perfil: {perfil}")
````

````python
# Combinar todos los tipos
def funcion_completa(a, b, *args, **kwargs):
    print(f"  a={a}, b={b}")
    print(f"  args={args}")
    print(f"  kwargs={kwargs}")

print("funcion_completa(1, 2):")
funcion_completa(1, 2)

print("
funcion_completa(1, 2, 3, 4, x=10, y=20):")
funcion_completa(1, 2, 3, 4, x=10, y=20)
````

### 4.3 Ámbito de variables (scope)

| Ámbito | Alcance |
|--------|---------|
| **Local** | Solo dentro de la función |
| **Global** | En todo el módulo |

Cuando queremos acceder a una variable global, deberemos anteponer la palabra reserbaga `global`si vamos a modificar el valor, de lo contrario python creará una nueva variable local con ese nombre. No será necesario cuando el acceso sea de solo lectura.

### 4.4 Funciones Lambda

Una función lambda es una función __anónima__  y de una sola expresión. Se usan para crear funciones pequeñas y rápidas sin necesidad de definirlas con def. Son particularmente útiles para operaciones simples como argumento de sorted(), filter(), map().

Sintaxis: `lambda parametros: expresion`

| Componente | Descripción |
|------------|-------------|
| **Parámetros** | Pueden ser cero o varios (separados por comas). |
| **Expresión** | La operación a realizar. Solo puede haber una. |
| **Retorno** | El resultado de la expresión se devuelve automáticamente (no hace falta `return`). |


````python
# Lambda basica
doble = lambda x: x * 2
print(f"doble(5) = {doble(5)}")
# Lambda con múltiples parametros
suma_lambda = lambda a, b: a + b
print(f"suma_lambda(3, 7) = {suma_lambda(3, 7)}")

# Usar lambda con sorted()
peliculas = [
    {"titulo": "Inception", "year": 2010, "rating": 8.8},
    {"titulo": "The Matrix", "year": 1999, "rating": 8.7},
    {"titulo": "Interstellar", "year": 2014, "rating": 8.6},
]

# Ordenar por año
ordenadas = sorted(peliculas, key=lambda p: p["year"])
print("Ordenadas por año:")
for p in ordenadas:
    print(f"  {p["year"]} - {p["titulo"]}")
````

````python
# Lambda con filter() - filtrar peliculas con rating >= 8.7
altas = filter(lambda p: p["rating"] >= 8.7, peliculas)
print("Peliculas con rating >= 8.7:")
for p in altas:
    print(f"  {p["titulo"]} ({p["rating"]})")

# Lambda con map() - extraer solo los títulos
titulos = map(lambda p: p["titulo"], peliculas)
print(f"
Títulos: {list(titulos)}")

# Lista de lambdas
operaciones = [
    lambda x: x + 1,
    lambda x: x * 2,
    lambda x: x ** 2,
]

print("Aplicar operaciones a 5:")
for op in operaciones:
    print(f"  {op(5)}")  # 6, 10, 25
````

#### 4.4.1 Comparativa __`lambda`__ vs __`def`__

| Característica | `lambda` | `def` |
|----------------|----------|-------|
| **Nombre** | Anónima (no tiene nombre) | Tiene nombre |
| **Cuerpo** | Una sola expresión | Bloque de código (varias líneas) |
| **Retorno** | Implícito (automático) | Explícito (`return`) |
| **Uso típico** | Funciones cortas, callbacks, `sort`, `map` | Lógica compleja, reutilizable |
| **Legibilidad** | Baja si es larga | Alta |

__NOTA__

>Llegados a este punto, pensaréis que las funciones lambda si tienen nombre. Al fin y al cabo lo estáis usando para invocarla. Pero esto no es correcto. El \"nombre\" con el que se invoca es la variable donde se almacena la referencia a la función. Probad a ejecutar el siguiente código:

````python
doble = lambda x: x * 2
print(f"Nombre de la función lambda: {doble.__name__}")

def calcular(x):
    return x * 2

print(f"Nombre de la funcion calcular: {calcular.__name__}")
````

>Como podéis ver, las funciones lamda se identifican siempre como \<lambda\>, lo cual implica una dificultad añadida cuando estemos llevando a cabo tareas de depuración, pues todas las funciones lambda se identificarán igual y será más dificil localizar el error.

## 5. Manejo de Excepciones

El manejo de excepciones permite que tu programa se recupere de situaciones
inesperadas sin abortar su ejecución. Es fundamental para programas robustos.

### 5.1 Bloques try, except, else, finally

| Bloque | Cuándo se ejecuta |
|--------|-------------------|
| try | Se intenta ejecutar código que puede fallar |
| except | Se ejecuta si ocurre una excepción |
| else | Se ejecuta si **no** hay excepción |
| finally | Se ejecuta **siempre**, haya o no excepción

````python
# try/except basico
print("Ejemplo 1: División segura")
def division_segura(a, b):
    try:
        resultado = a / b
        print(f"  {a} / {b} = {resultado}")
    except ZeroDivisionError:
        print(f"  Error: no se puede dividir {a} entre 0")

division_segura(10, 2)   # -> 5.0
division_segura(10, 0)   # -> Error: no se puede dividir 10 entre 0

# try/except con multiples excepciones
print("Ejemplo 2: Multiples excepciones")

def procesar_dato(valor):
    try:
        numero = int(valor)
        print(f"  '{valor}' -> entero: {numero}")
    except ValueError:
        print(f"  '{valor}' -> no es un número entero válido")

procesar_dato("42")      # -> 42
procesar_dato("hola")    # -> no es un número entero válido
procesar_dato("3.14")    # -> no es un número entero válido
````

````python
# try/except/else/finally
print("Ejemplo 3: try/except/else/finally")

def abrir_archivo_simulado(nombre):
    try:
        print(f"  Intentando abrir '{nombre}'...")
        if nombre == "inventario.csv":
            return {"productos": ["manzana", "platan"]}
        else:
            raise FileNotFoundError(f"'{nombre}' no existe")
    except FileNotFoundError:
        print(f"  Error: archivo no encontrado")
    except TypeError:
        print(f"  Error: tipo de dato incorrecto")
    else:
        print(f"  Archivo abierto correctamente")
        return None
    finally:
        print(f"  Cerrando recursos de '{nombre}'")

# Caso exitoso
print("--- Caso exitoso ---")
resultado = abrir_archivo_simulado("inventario.csv")
print(f"  Resultado: {resultado}")

# Caso con error
print("--- Caso con error ---")
resultado = abrir_archivo_simulado("datos.csv")
print(f"  Resultado: {resultado}")
````

### 5.2 Excepciones comunes

| Excepción | Cuándo ocurre |
|-----------|---------------|
| ValueError | Valor correcto pero tipo incorrecto |
| TypeError | Operación con tipo incorrecto |
| IndexError | Índice fuera de rango |
| KeyError | Clave no existe en diccionario |
| FileNotFoundError | Archivo no encontrado |
| ZeroDivisionError | División entre cero |
| AttributeError | Objeto no tiene el atributo

````python
# Ejemplo: manejo de diferentes excepciones
def procesar_datos(datos):
    try:
        # Intentar acceder a un diccionario
        valor = datos["precio"]
        
        # Intentar convertir a número
        precio = float(valor)
        
        # Intentar operar
        if precio < 0:
            raise ValueError("El precio no puede ser negativo")
        
        iva = precio * 1.21
        print(f"  Precio con IVA: {iva:.2f}")
        
    except KeyError:
        print("  Error: la clave precio no existe en los datos")
    except (ValueError, TypeError) as e:
        print(f"  Error de valor: {e}")
    except Exception as e:
        print(f"  Error inesperado: {type(e).__name__}: {e}")

# Casos de prueba
procesar_datos({"precio": "100"})       # -> Precio con IVA: 121.00
procesar_datos({"precio": "abc"})       # -> Error de valor
procesar_datos({"nombre": "Producto"})  # -> Error: clave no existe
procesar_datos({"precio": -50})         # -> Error: precio negativo
````

### 5.3 Lanzar excepciones con raise

### 6.2 Archivos CSV con el modulo csv

Los archivos CSV (Comma-Separated Values) son el formato mas comun para
datos tabulares. Python incluye el modulo csv para trabajar con ellos.

````python
import csv

# Escribir un CSV
datos = [
    ["Nombre", "Edad", "Ciudad"],
    ["Ana", 25, "Madrid"],
    ["Carlos", 30, "Barcelona"],
    ["Beatriz", 22, "Valencia"],
]

with open("datos.csv", "w", encoding="utf-8", newline="") as archivo:
    escritor = csv.writer(archivo)
    escritor.writerows(datos)

print("Archivo datos.csv creado.")

# Leer un CSV
print("Contenido de datos.csv:")
with open("datos.csv", "r", encoding="utf-8") as archivo:
    lector = csv.reader(archivo)
    for fila in lector:
        print(f"  {fila}")
````

````python
# Leer CSV como diccionarios (encabezados como claves)
print("CSV como diccionarios:")
with open("datos.csv", "r", encoding="utf-8") as archivo:
    lector = csv.DictReader(archivo)
    for fila in lector:
        print(f"  {fila["Nombre"]} tiene {fila["Edad"]} años, vive en {fila["Ciudad"]}")
````

### 6.3 Importacion de modulos

Python viene con **modulos estandar** (biblioteca estandar) y puedes
importar los tuyos propios.

````python
# Diferentes formas de importar
import math
import os
from datetime import datetime
from collections import Counter, defaultdict

# math: funciones matematicas
print(f"PI = {math.pi:.4f}")
print(f"raiz(2) = {math.sqrt(2):.4f}")
print(f"2^10 = {math.pow(2, 10):.0f}")
print(f"ceil(3.2) = {math.ceil(3.2)}")
print(f"floor(3.8) = {math.floor(3.8)}")

# os: operaciones con sistema de archivos
print(f"Directorio actual: {os.getcwd()}")

# datetime: fechas y horas
ahora = datetime.now()
print(f"Fecha/hora actual: {ahora.strftime("%Y-%m-%d %H:%M:%S")}")

# collections: estructuras de datos avanzadas
palabras = ["manzana", "platano", "manzana", "pera", "platano", "manzana"]
contador = Counter(palabras)
print(f"Contador de palabras: {contador}")
print(f"Más común: {contador.most_common(1)[0]}")
````

### 6.4 Trabajando con rutas de archivos

Usa el modulo pathlib (recomendado) o os.path para manejar rutas.

````python
from pathlib import Path

# Crear una ruta
ruta = Path(".")

# Listar archivos Python
print("Archivos Python en directorio:")
for archivo in ruta.glob("*.py"):
    print(f"  {archivo.name} ({archivo.stat().st_size} bytes)")

# Verificar existencia
print(f"ejemplo.txt existe: {Path("ejemplo.txt").exists()}")
print(f"no_existe.txt existe: {Path("no_existe.txt").exists()}")

# Crear directorios
directorio = Path("carpeta_ejemplo")
directorio.mkdir(exist_ok=True)
print(f"Directorio creado: {directorio.absolute()}")

# Unir rutas
ruta_completa = Path("carpeta_ejemplo") / "archivo.txt"
print(f"Ruta completa: {ruta_completa}")
print(f"Extension: {ruta_completa.suffix}")  # .txt
print(f"Nombre sin extension: {ruta_completa.stem}")  # archivo

# Limpiar
import shutil
shutil.rmtree("carpeta_ejemplo")
print("Directorio de ejemplo eliminado.")
````
